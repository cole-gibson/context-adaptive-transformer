## utilities

general

```python
def get_state_idx(config, path, t: int):
    """Get state idx for state at time t' <= t"""
    state_iters = np.array(config.log.state.state_iters)
    state_idx = len(state_iters[state_iters < t])-1

    path = Path(path)
    for idx in range(state_idx+1, 1000):
        test_path = path / f'{idx}.pt'
        if test_path.exists():
            test_state_dict = torch.load(test_path)
            s = test_state_dict['iter']
            if s > t:
                return idx-1
                break
        else:
            print(f"State file {test_path} does not exist. Returning last valid index {idx-1}.")
            return idx-1
            break

def filter_t(df, t_value, t_col='t'):
    """
    Returns a filtered DataFrame with rows where s_col is the closest value ≤ s_value.
    If all values are greater than s_value, returns the smallest s_col instead.
    """
    if t_value is None:
        raise ValueError("You must provide s_value.")

    # Filter to those <= s_value if possible
    subset = df[df[t_col] <= t_value]
    if subset.empty:
        # fallback: take the smallest s
        subset = df

    # find the row(s) with the closest s
    closest_val = subset[t_col].max()
    return closest_val, df[df[t_col] == closest_val]

def mask_df(df: pd.DataFrame, exp_config, seed_idx, n_idx, param_idx, param_name: str = None):
    if param_name is None:
        param_name == exp_config.param_name
    return df[(df['seed'] == exp_config.seedrange[seed_idx]) & (df['n'] == exp_config.nrange[n_idx]) & (df[param_name] == exp_config.paramrange[param_idx])]

def process_edge_causal(edge_causal: pd.DataFrame):
    names = {
        '0': 'tok',
        '1': 'a1',
        '2': 'm1',
        '3': 'a2',
        '4': 'm2'
    }

    full_loss = edge_causal[edge_causal['n_edges'] == 10].copy(deep = True)
    out = edge_causal[edge_causal['n_edges'] == 9].copy(deep = True)
    out['str_disabled_edges'] = out['disabled_edges'].apply(lambda x: names[x[2]] + '_' + names[x[5]])
    return full_loss, out

def name(seed, k, n):
    return f'S{seed}_K{k}_N{n}'

def get_params(exp_config, seed_idx, param_idx, n_idx):
    seed = exp_config.seedrange[seed_idx]
    k = exp_config.paramrange[param_idx]
    n = exp_config.nrange[n_idx]

    return seed, k, n

def load_config(base_dir: str, seed_idx: int, k_idx: int, n_idx: int):
    base_dir = Path(base_dir)
    exp_config = u.load_config(base_dir / 'exp_config.yaml')
    seed, k, n = get_params(exp_config, seed_idx, k_idx, n_idx)
    config = u.load_config_and_task_pool(base_dir / 'data' / name(seed, k, n))
    
    return config

def load_model(base_dir: str, seed_idx: int, k_idx: int, n_idx: int):
    config = load_config(base_dir, seed_idx, k_idx, n_idx)
    get_model = u.dynamic_import(exp_config.model, 'get_model_safe')
    
    return get_model(config)

def load_task_model(base_dir: str, seed_idx: int, k_idx: int, n_idx: int):
    base_dir = Path(base_dir)
    exp_config = u.load_config(base_dir / 'exp_config.yaml')
    config = load_config(base_dir, seed_idx, k_idx, n_idx)
    get_task_model = u.dynamic_import(exp_config.task_model, 'get_task_model_safe')
    
    return get_task_model(config)

def load_all(base_dir: str, seed_idx: int, k_idx: int, n_idx: int):
    base_dir = Path(base_dir)
    return load_config(base_dir, seed_idx, k_idx, n_idx), load_model(base_dir, seed_idx, k_idx, n_idx), load_task_model(base_dir, seed_idx, k_idx, n_idx)

def get_state_dict(base_dir: str, seed_idx: int, k_idx: int, n_idx: int, state: int = None):
    base_dir = Path(base_dir)
    exp_config = u.load_config(base_dir / 'exp_config.yaml')
    seed, k, n = get_params(exp_config, seed_idx, k_idx, n_idx)
    state_dir = base_dir / 'data' / name(seed, k, n) / 'state'
    if state is None:
        pt_files = [
            f for f in state_dir.glob("*.pt")
            if f.stem.isdigit()
        ]
        state = max(int(f.stem) for f in pt_files)
    load = torch.load(state_dir / f'{state}.pt')
    iter, state_dict = load['iter'], load['state']
    return iter, state_dict
```

```python
def ablation_losses(
    model: nn.Module,
    x: torch.Tensor,
    y: torch.Tensor,
    layers: List[nn.Module],
    max_order: int = 1,
    ablation_type: str = 'zero',  # 'zero' or 'mean'
) -> Dict[Tuple[int, ...], float]:
    """
    Computes the loss for all ablations of 0..max_order layers.

    Args:
        model:         your Transformer‐like model
        x, y:          input and target tensors
        loss_fn:       e.g. nn.CrossEntropyLoss()
        layers:        list of the N layer‐modules to consider ablating
        max_order:     maximum size of layer‐sets to ablate
        ablation_type: 'zero' to zero-out layer outputs,
                       'mean' to replace outputs with batch-mean (broadcasted)

    Returns:
        A dict mapping each tuple of layer‐indices to the scalar loss.
        The key `()` is the control (no layers ablated).
    """
    device = x.device
    results: Dict[Tuple[int, ...], float] = {}

    # Define hooks
    def zero_hook(module, inp, out):
        return torch.zeros_like(out)

    def mean_hook(module, inp, out):
        # compute mean over batch dim, keep all other dims
        mean = out.mean(dim=0, keepdim=True)
        return mean.expand_as(out)

    # pick hook
    if ablation_type == 'zero':
        make_hook = lambda: zero_hook
    elif ablation_type == 'mean':
        make_hook = lambda: mean_hook
    else:
        raise ValueError(f"Unknown ablation_type: {ablation_type!r}")

    # --- Control (no ablation) ---
    model.zero_grad()
    _, loss = model(x, y)
    results[()] = loss.item()

    # --- Ablations ---
    for k in range(1, max_order + 1):
        for combo in combinations(range(len(layers)), k):
            # register hooks
            hooks = [layers[idx].register_forward_hook(make_hook()) for idx in combo]

            # forward + record
            model.zero_grad()
            _, loss = model(x, y)
            results[combo] = loss.item()

            # clean up
            for h in hooks:
                h.remove()

    return results
```

### loss

```python
def plot_loss(cur_ax, model_perf, est_perf, exp_config,
                  est_idx_list: list = list(range(4)),
                  dist: list = ['train', 'val'],
                  alpha_val: float = 0.5,
                  plot_kwargs: dict = {}):

    param_name = exp_config.param_name

    # --- estimator lines first ---
    data = est_perf
    palette = sns.color_palette()

    legend_colors = [palette[1], palette[2], palette[3], palette[0]]
    legend_names = ['1-Mem', '1-Gen', '2-Mem', '2-Gen']
    color_dict = {name: color for name, color in zip(legend_names, legend_colors)}

    est_names = ['1_mem', '1_gen', '2_mem', '2_gen']

    legend_handles = []
    for i in est_idx_list:
        for d in dist:
            est = est_names[i]
            mask = (data['name'] == est) & (data['dist'] == d)
            color = color_dict[legend_names[i]]
            cur_ax.axhline(
                y=data[mask]['loss'].unique().item(),
                label=f'{est}',
                color=color,
                alpha=alpha_val if d == 'val' else 1.0,
                **plot_kwargs
            )

        legend_handles.append(
            mlines.Line2D([], [], color=color, linestyle='-', label=f'{est}')
        )

    # --- model curves last (on top) ---
    data = model_perf
    color = 'black'

    for d in dist:
        mask = data['dist'] == d
        sns.lineplot(
            data=data[mask],
            x='t',
            y='loss',
            label=d,
            ax=cur_ax,
            color=color,
            alpha=alpha_val if d == 'val' else 1.0,
            **plot_kwargs
        )
    
    legend_handles.append(
        mlines.Line2D([], [], color=color, linestyle='-', label='Model', linewidth=1)
    )

    cur_ax.get_legend().remove()

    cur_ax.set_title(rf'${param_name}={data[param_name].unique().item()}$')
    cur_ax.set_xlabel(r'$t$')
    cur_ax.set_ylabel('Loss')

    return legend_handles
```

```python
df = edge_causal.copy(deep = True)
mask = (df['ablated_edge'] != '(-1, -1)') & (df['ablated_edge'] != '()')
df = df[mask]

    # 0:        token                           0
    # 1, 0:     attention 1_Q    dest           1
    # 1, 1:     attention 1_K    dest           2
    # 1, 2:     attention 1_V    source         3
    # 2:        MLP                             4
    # 3, 0:     attention 2_Q    dest           5
    # 3, 1:     attention 2_K    dest           6
    # 3, 2, 0:  attention 2_V_0  ablate source  7
    # 3, 2, 1:  attention 2_V_1  ablate source  8
    # 3, 2, 2:  attention 2_V_2  ablate source  9
    # 3, 2:     attention 2_V    source         10
    # 4:        MLP                             11
    # 5:        unembed                         12

names = {
    0: 'Tok',
    1: 'Att1Q',
    2: 'Att1K',
    3: 'Att1V',
    4: 'MLP1',
    5: 'Att2Q',
    6: 'Att2K',
    7: 'Att2VTok',
    8: 'Att2VAtt1',
    9: 'Att2VMLP1',
    11: 'MLP2',
    12: 'Lin'
}

df['ablated_edge'] = df['ablated_edge'].apply(lambda x: ast.literal_eval(x))
df['from'] = df['ablated_edge'].apply(lambda x: names[x[0]])
df['to'] = df['ablated_edge'].apply(lambda x: names[x[1]])

# set ordering for convenience
to_order = ['Lin','MLP2','Att2K','Att2Q','MLP1', 'Att1K','Att1Q']
from_order = ['Tok','Att1V','MLP1','Att2VTok','Att2VAtt1','Att2VMLP1','MLP2','Lin']

df['from'] = pd.Categorical(df['from'], categories = from_order, ordered=True)
df['to'] = pd.Categorical(df['to'], categories=to_order, ordered=True)

df['logK'] = df['K'].apply(lambda x: np.log2(x))
# derivative wrt to t of 'kl_norm' grouping by seed, K, n, and ablated_edge
df['d_kl_norm'] = df.groupby(['seed', 'K', 'n', 'ablated_edge'])['kl_norm'].diff().fillna(0)
```

```python
def circuit_plot(ax, t, K, name: str, show_xlabels=False, derivative=False):
    t, filtered_df = filter_t(df, t)
    mask = (filtered_df['K'] == K)
    if derivative:
        snapshot = filtered_df[mask].pivot_table(values='d_kl_norm', index='from', columns='to', observed=False)
        v = np.max(np.abs(snapshot))
        sns.heatmap(snapshot, cmap='vlag', center=0, vmin=-v, vmax=v, square=True, ax = ax, cbar=True)
    else:
        snapshot = filtered_df[mask].pivot_table(values='kl_norm', index='from', columns='to', observed=False)
        sns.heatmap(snapshot, cmap='Grays', vmin=0, vmax = 1, square=True, ax = ax, cbar=True, annot=False, fmt=".1f", annot_kws={"size": 6})
    ax.set_yticklabels(['State', 'Att1(State)', 'MLP1', 'Att2(State)', 'Att2(Att1)', 'Att2(MLP1)', 'MLP2'], rotation=0)
    ax.set_xticklabels(['Linear', 'MLP2', 'Att2K', 'Att2Q', 'MLP1', 'Att1K', 'Att1Q'], rotation=90)
    ax.tick_params(axis='both', which='both', length=0)
    ax.set_xlabel('')
    ax.set_ylabel('')
    ax.set_title(name)
    ax.invert_xaxis()
    # cbar = ax.collections[0].colorbar
    # cbar.set_label(r"$D_{KL}(\hat\pi | \tilde\pi)$")

    if not show_xlabels:
        ax.set_xticklabels([])
        ax.set_xlabel("")

    # draw boxes
    data = snapshot.to_numpy()
    rows, cols = np.where(~np.isnan(data))
    for r, c in zip(rows, cols):
        ax.add_patch(Rectangle(
            (c, r),       # (x, y) in data coordinates (col, row)
            1, 1,
            fill=False,
            edgecolor="black",
            linewidth=0.5
        ))
```

```python
from base.operations.toolkit import get_res_act

def get_tv(param_idx, seed_idx, n_idx, t, B = 2**10):
    K = exp_config.paramrange[param_idx]

    get_model_safe = u.dynamic_import(exp_config.model, 'get_model_safe')
    get_task_model_safe = u.dynamic_import(exp_config.task_model, 'get_task_model_safe')
    model_name = naming(exp_config.seedrange[seed_idx], exp_config.nrange[n_idx], exp_config.param_name, K)

    config = u.load_config_and_task_pool(Path(exp_config.base_dir) / 'data' / model_name)
    model = get_model_safe(config)
    task_model = get_task_model_safe(config)
    model.eval()
    state_path = Path(exp_config.base_dir) / 'data' / model_name / 'state'
    state_idx = get_state_idx(config, state_path, t)
    state_dict = torch.load(state_path / f'{state_idx}.pt')
    s = state_dict['iter']
    print(f"Loaded state dict from iter {s} (t={t})")
    state_dict = state_dict['state']
    model.load_state_dict(state_dict)
    x, y, tasks = task_model.get_batch(B, config.training.context_len, return_tasks=True)
    xx, _ = task_model.get_batch(B, config.training.context_len, dist='val')
    task_pool = task_model.task_pool
    tv = get_res_act(model, (x, y))

    return tv, tasks[:, -1], task_model
```

```python
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

def tsne(data, labels, title, ax):
    # Convert to numpy
    x_np = data.detach().cpu().numpy()
    labels_np = labels.detach().cpu().numpy()

    # Run t-SNE
    tsne = TSNE(
        n_components=2,
        perplexity=30.0,
        learning_rate="auto",
        init="pca",
        random_state=0,
    )
    print(x_np.shape)
    print('beginning t-SNE...')
    x_2d = tsne.fit_transform(x_np)   # (B, 2)

    print('t-SNE completed, building plot...')
    # Build dataframe for seaborn
    df = pd.DataFrame({
        "tsne_1": x_2d[:, 0],
        "tsne_2": x_2d[:, 1],
        "label": labels_np.astype(int)
    })
    print('DataFrame for plotting created, plotting...')
    sns.scatterplot(
        data=df,
        x="tsne_1",
        y="tsne_2",
        hue="label",
        palette="tab20",
        s=3,
        alpha=0.8,
        edgecolor=None,
        ax = ax
    )
    
    ax.text(0.05, 0.05, title, ha='left', va='bottom', transform=ax.transAxes)
    ax.legend().remove()
    ax.set_xlabel('')
    ax.set_ylabel('')
    ax.set_xticks([])
    ax.set_yticks([])
```

```python
def tv_tsne(param_idx, seed_idx, n_idx, t, name, ax, B = 2**10):
    u.set_seed(10)
    tv, tasks, _ = get_tv(param_idx, seed_idx, n_idx, t, B = B)
    data = tv[2][:, -1, :]
    tsne(data, tasks, name, ax)
```

#### patching

```python
def accuracy(logits1, logits2):
    assert logits1.shape == logits2.shape
    n = logits1.shape[0]
    p1 = F.softmax(logits1, -1)
    p2 = F.softmax(logits2, -1)
    _, idx1 = p1.max(-1)
    _, idx2 = p2.max(-1)

    return (torch.sum(idx1 == idx2)/n).item()

def similarity(logits1, logits2):
    assert logits1.shape == logits2.shape
    
    return math.exp(-u.KL(logits1, logits2)/u.KL(logits1, torch.zeros_like(logits1)))
    # return 1 - KL(logits1, logits2)/KL(logits1, torch.zeros_like(logits1))
```

```python
from pd.compute import naming
from base.estimators.markov_model_ar import BiMem, UniMem
from base.operations.toolkit import get_res_act, patched_res_act_performance
from matplotlib import patches as mpatches

def perturb_plot(ax, seed_idx, n_idx, param_idx, t, name: str, estimator: str, attn_idx: int, minimal = False):
    u.set_seed(10)
    K = exp_config.paramrange[param_idx]

    get_model_safe = u.dynamic_import(exp_config.model, 'get_model_safe')
    get_task_model_safe = u.dynamic_import(exp_config.task_model, 'get_task_model_safe')
    model_name = naming(exp_config.seedrange[seed_idx], exp_config.nrange[n_idx], exp_config.param_name, K)

    config = u.load_config_and_task_pool(Path(exp_config.base_dir) / 'data' / model_name)
    model = get_model_safe(config)
    task_model = get_task_model_safe(config)
    model.eval()
    state_path = Path(exp_config.base_dir) / 'data' / model_name / 'state'
    state_idx = get_state_idx(config, state_path, t)
    state_dict = torch.load(state_path / f'{state_idx}.pt')
    s = state_dict['iter']
    state_dict = state_dict['state']
    model.load_state_dict(state_dict)
    B = 2**10
    N = exp_config.nrange[n_idx]
    xa, ya, tasks = task_model.get_batch(B, config.training.context_len, return_tasks=True)
    xb, yb = task_model.get_batch(B, config.training.context_len)
    if estimator == '2_mem':
        estimator = BiMem(config, task_model.task_pool.float(), task_model.stat_dist.float())
    elif estimator == '1_mem':
        estimator = UniMem(config, task_model.task_pool.float(), task_model.stat_dist.float())
    
    # (x, y) = posterior on x, predict on current states of y
    behavior = {}
    behavior['AA'], _, hatTa = estimator(xa, ya, return_hatT=True)
    behavior['BB'], _, hatTb = estimator(xb, yb, return_hatT=True)
    behavior['AB'] = torch.log(torch.einsum('ijkl,ijk->ijl', hatTa, F.one_hot(xb, config.task.vocab_size).float()))
    behavior['BA'] = torch.log(torch.einsum('ijkl,ijk->ijl', hatTb, F.one_hot(xa, config.task.vocab_size).float()))

    a_logits, _ = model(xa, ya)
    b_logits, _ = model(xb, yb)
    a_tv = get_res_act(model, (xa, ya))
    b_tv = get_res_act(model, (xb, yb))

    output = {
        'predictor': [],
        'accuracy': [],
        'condition': [],
        'similarity': [],
    }
    
    bpa_logits, _ = patched_res_act_performance(model, [attn_idx], [a_tv[attn_idx]], (xb, yb))
    apb_logits, _ = patched_res_act_performance(model, [attn_idx], [b_tv[attn_idx]], (xa, ya))
    for key in behavior.keys():
        t = behavior[key]

        output['similarity'].append(similarity(t.flatten(0,1), a_logits))
        output['accuracy'].append(accuracy(t.flatten(0,1), a_logits))
        output['predictor'].append(key)
        output['condition'].append('SeqA')

        output['similarity'].append(similarity(t.flatten(0,1), b_logits))
        output['accuracy'].append(accuracy(t.flatten(0,1), b_logits))
        output['predictor'].append(key)
        output['condition'].append('SeqB')

        output['similarity'].append(similarity(t.flatten(0,1), bpa_logits))
        output['accuracy'].append(accuracy(t.flatten(0,1), bpa_logits))
        output['predictor'].append(key)
        output['condition'].append('SeqB (Patch A)')

        output['similarity'].append(similarity(t.flatten(0,1), apb_logits))
        output['accuracy'].append(accuracy(t.flatten(0,1), apb_logits))
        output['predictor'].append(key)
        output['condition'].append('SeqA (Patch B)')
    
    output = pd.DataFrame(output)

    baseline_similarity = similarity(t.flatten(0,1), torch.zeros_like(apb_logits))  # unconditioned baseline

    if minimal:
        plot_data = output[output['condition'].isin(['SeqA', 'SeqB (Patch A)']) & output['predictor'].isin(['AA', 'AB'])].copy()
    else:
        plot_data = output.copy()

    if minimal:
        hue_order = ['SeqA', 'SeqB (Patch A)']
        hatches = ['', '']
    else:
        hue_order = ['SeqA', 'SeqB', 'SeqB (Patch A)', 'SeqA (Patch B)']
        hatches = ['/', '\\\\', 'xx', '..']

    sns.barplot(
        data=plot_data,
        x='predictor',
        y='similarity',
        hue='condition',
        hue_order=hue_order,
        ax=ax,
        legend=True,
        palette=['C7', 'white'],
        edgecolor='black',
    )

    # for i, bar in enumerate(ax.patches):
    #     bar.set_facecolor('white' if i // 2 ==1 else 'black')
    #     bar.set_edgecolor('black')
    #     # bar.set_hatch(hatches[i % len(hue_order)])

    # legend_handles = [
    #     mpatches.Patch(facecolor='black', edgecolor='white', hatch=h, label=label)
    #     for h, label in zip(hatches, hue_order)
    # ]
    # ax.legend(handles=legend_handles, title='condition')
    # sns.barplot(data = plot_data, x = 'predictor', y = 'similarity', hue='condition', ax = ax, legend=True, palette=['C6', 'C9', 'C6', 'C9'])
    # ax.set_title(f'{name} Task Vector Patching')
    ax.set_xlabel('Predictor')
    ax.set_ylabel('Similarity')
    # ax.axhline(baseline_similarity, color='black', linestyle=':', label='Uniform Baseline')
    
    return output, behavior, ax, s
```

```python
from matplotlib.colors import LinearSegmentedColormap, to_rgba
from matplotlib import colormaps

def black_to_color(name=None, final_color=None, N=256, reverse=True):
    """
    Return a colormap that goes between black and a target color.

    Parameters
    ----------
    name : str, optional
        Name of a Matplotlib colormap (e.g. 'Reds'). Used if final_color is None.
    final_color : str or tuple, optional
        The final color (matplotlib-recognized format). If None, take from colormap.
    N : int
        Number of levels in the colormap.
    reverse : bool
        If True, make the gradient go from the target color to black.
    """
    if final_color is None:
        if name is None:
            raise ValueError("Either `name` or `final_color` must be provided.")
        base = colormaps.get_cmap(name)
        target = base(1.0)  # brightest color
    else:
        target = to_rgba(final_color)  # convert to RGBA

    colors = [(0, 0, 0, 1), target]
    if reverse:
        colors = colors[::-1]

    cmap = LinearSegmentedColormap.from_list(
        (name or "custom") + "_black" + ("_rev" if reverse else ""),
        colors,
        N=N
    )
    return cmap

def log_heatmap(pvt, ax=None, vmin=None, vmax=None, cmap="viridis", rasterized=True):
    if ax is None:
        ax = plt.gca()

    X = np.asarray(pvt.columns, dtype=float)
    Y = np.asarray(pvt.index, dtype=float)
    Z = pvt.values

    def log_edges(vals):
        vals = np.asarray(vals, float)
        ratios = vals[1:] / vals[:-1]
        edges = np.empty(len(vals) + 1)
        edges[1:-1] = np.sqrt(vals[:-1] * vals[1:])
        edges[0] = vals[0] / np.sqrt(ratios[0])
        edges[-1] = vals[-1] * np.sqrt(ratios[-1])
        return edges

    x_edges = log_edges(X)
    y_edges = log_edges(Y)

    qm = ax.pcolormesh(
        x_edges, y_edges, Z,
        shading="auto",
        vmin=vmin, vmax=vmax, cmap=cmap,
        edgecolors="none",
        linewidth=0,
        antialiased=False,
        rasterized=rasterized,   # <-- key
    )

    ax.set_xscale("log")
    ax.set_yscale("log", base=2)

    for spine in ax.spines.values():
        spine.set_visible(False)

    return ax, qm

import numpy as np
import matplotlib.pyplot as plt

def composite_cmap_colorbar(
    fig,
    colors,
    *,
    vmin,
    vmax,
    ax_rect=(1.05, 0.2, 0.025, 0.7),   # (x, y, w_per_cmap, h) in figure coords
    n=256,                             # vertical resolution
    ticks=None,
    tick_right=True,
    label="",
    remove_spines=True,
    separator=True,
    separator_kwargs=None,
    cmap_factory=None,                 # function(color)->cmap; defaults to black_to_color(final_color=color)
):
    """
    Draw a composite vertical "colorbar" consisting of adjacent columns, one per colormap,
    all sharing the same [vmin, vmax] normalization.

    Parameters
    ----------
    fig : matplotlib.figure.Figure
    colors : sequence
        Sequence of colors (any matplotlib-acceptable color objects).
    vmin, vmax : float
    ax_rect : (x, y, w_per_cmap, h)
        Rectangle in figure coordinates. Width is per-colormap; total width scales with len(colors).
    n : int
        Number of samples along the vertical axis.
    ticks : sequence or None
        Y ticks in data units. If None, no ticks are set.
    tick_right : bool
        Place y ticks on the right.
    label : str
        Optional y-axis label (acts like a colorbar label).
    remove_spines : bool
        Hide all spines.
    separator : bool
        Draw vertical separators between columns.
    separator_kwargs : dict or None
        Passed to ax.vlines; defaults to thin white separators.
    cmap_factory : callable or None
        If provided, called as cmap_factory(color) -> matplotlib Colormap.
        If None, uses black_to_color(final_color=color).

    Returns
    -------
    ax : matplotlib.axes.Axes
    """
    if cmap_factory is None:
        def cmap_factory(c):
            return black_to_color(final_color=c)

    x, y, w_per, h = ax_rect
    k = len(colors)
    if k == 0:
        raise ValueError("colors must be a non-empty sequence.")

    ax = fig.add_axes([x, y, w_per * k, h])

    values = np.linspace(vmin, vmax, n)[:, None]  # (n, 1)
    norm = plt.Normalize(vmin=vmin, vmax=vmax)

    for i, c in enumerate(colors):
        ax.imshow(
            values,
            cmap=cmap_factory(c),
            norm=norm,
            aspect="auto",
            origin="lower",
            extent=(i, i + 1, vmin, vmax),
        )

    ax.set_xlim(0, k)
    ax.set_ylim(vmin, vmax)
    ax.set_xticks([])

    if ticks is not None:
        ax.set_yticks(list(ticks))

    if tick_right:
        ax.yaxis.tick_right()
        ax.yaxis.set_label_position("right")

    if label:
        ax.set_ylabel(label)

    if remove_spines:
        for spine in ax.spines.values():
            spine.set_visible(False)

    if separator and k > 1:
        if separator_kwargs is None:
            separator_kwargs = dict(colors="white", linewidth=0.5)
        ax.vlines(np.arange(1, k), vmin, vmax, **separator_kwargs)

    return ax
```

### loss

```python
def plot_loss(cur_ax, model_perf, est_perf, exp_config,
                  est_idx_list: list = list(range(4)),
                  dist: list = ['train', 'val'],
                  alpha_val: float = 0.5,
                  plot_kwargs: dict = {}):

    param_name = exp_config.param_name

    # --- estimator lines first ---
    data = est_perf
    palette = sns.color_palette()

    legend_colors = [palette[1], palette[2], palette[3], palette[0]]
    legend_names = ['1-Mem', '1-Gen', '2-Mem', '2-Gen']
    color_dict = {name: color for name, color in zip(legend_names, legend_colors)}

    est_names = ['1_mem', '1_gen', '2_mem', '2_gen']

    legend_handles = []
    for i in est_idx_list:
        for d in dist:
            est = est_names[i]
            mask = (data['name'] == est) & (data['dist'] == d)
            color = color_dict[legend_names[i]]
            cur_ax.axhline(
                y=data[mask]['loss'].unique().item(),
                label=f'{est}',
                color=color,
                alpha=alpha_val if d == 'val' else 1.0,
                **plot_kwargs
            )

        legend_handles.append(
            mlines.Line2D([], [], color=color, linestyle='-', label=f'{est}')
        )

    # --- model curves last (on top) ---
    data = model_perf
    color = 'black'

    for d in dist:
        mask = data['dist'] == d
        sns.lineplot(
            data=data[mask],
            x='t',
            y='loss',
            label=d,
            ax=cur_ax,
            color=color,
            alpha=alpha_val if d == 'val' else 1.0,
            **plot_kwargs
        )
    
    legend_handles.append(
        mlines.Line2D([], [], color=color, linestyle='-', label='Model')
    )

    cur_ax.get_legend().remove()

    cur_ax.set_title(rf'${param_name}={data[param_name].unique().item()}$')
    cur_ax.set_xlabel(r'$t$')
    cur_ax.set_ylabel('Loss')

    return legend_handles
```

```python
# numerical derivative wrt to column s
df = model_perf[model_perf["dist"]=="val"].copy()
df['d_val'] = df['loss'].diff().fillna(0)
mask = (df['loss'] < 2.2) & (df['d_val'] > 1e-3)

result = (
    df[mask]
    .groupby(['seed', 'K', 'n'])['t']
    .min()
    .reset_index()
    .rename(columns={'t': 't_trans'})
)

ax = sns.violinplot(result, x='K', y='t_trans', native_scale=True, cut=0)
ax.set_xscale('log', base=2)
ax.set_yscale('log', base=10)
```

```python
keys = ['seed', 'K', 'n', 'dist']
df = model_perf.merge(
    est_perf[keys + ['loss', 'role', 'name']],
    on=keys,
    how='left',
    suffixes=('_model', '_est')
)
```

```python
wide = df[df['name_est']=='2_gen'].pivot_table(
    index=['seed', 'K', 'n', 't'],
    columns='dist',
    values=['loss_est', 'loss_model']
)
mask = (wide['loss_model', 'val'] < 2.3) & (wide['loss_model', 'train'] < wide['loss_est', 'train'])

result = (
    wide[mask]
    .reset_index()
    .groupby(['seed', 'K', 'n'])['t']
    .min()
    .reset_index()
    .rename(columns={'t': 't_trans'})
)

ax = sns.violinplot(result, x='K', y='t_trans',
                    native_scale=True, cut=1)

ax.set_xscale('log', base=10)
ax.set_yscale('log', base=10)
ax.set_ylabel(r'$\tau_\mathrm{transience}$')
ax.set_xlabel(r'$K$')

# --- linear fit in log–log space ---
x = np.asarray(result['K'])
y = np.asarray(result['t_trans'])

mask = (x > 0) & (y > 0)
logx = np.log10(x[mask])
logy = np.log10(y[mask])

slope, intercept = np.polyfit(logx, logy, 1)

xfit = np.logspace(logx.min(), logx.max(), 200)
yfit = 10 ** (intercept + slope * np.log10(xfit))

ax.plot(
    xfit, yfit,
    color='k', lw=1,
    linestyle=':',
    label=rf'$\propto K^{{{slope:.2f}}}$'
)

ax.legend()
```

```python
df = edge_causal.copy(deep = True)
mask = (df['ablated_edge'] != '(-1, -1)') & (df['ablated_edge'] != '()')
df = df[mask]

    # 0:        token                           0
    # 1, 0:     attention 1_Q    dest           1
    # 1, 1:     attention 1_K    dest           2
    # 1, 2:     attention 1_V    source         3
    # 2:        MLP                             4
    # 3, 0:     attention 2_Q    dest           5
    # 3, 1:     attention 2_K    dest           6
    # 3, 2, 0:  attention 2_V_0  ablate source  7
    # 3, 2, 1:  attention 2_V_1  ablate source  8
    # 3, 2, 2:  attention 2_V_2  ablate source  9
    # 3, 2:     attention 2_V    source         10
    # 4:        MLP                             11
    # 5:        unembed                         12

names = {
    0: 'Tok',
    1: 'Att1Q',
    2: 'Att1K',
    3: 'Att1V',
    4: 'MLP1',
    5: 'Att2Q',
    6: 'Att2K',
    7: 'Att2VTok',
    8: 'Att2VAtt1',
    9: 'Att2VMLP1',
    11: 'MLP2',
    12: 'Lin'
}

df['ablated_edge'] = df['ablated_edge'].apply(lambda x: ast.literal_eval(x))
df['from'] = df['ablated_edge'].apply(lambda x: names[x[0]])
df['to'] = df['ablated_edge'].apply(lambda x: names[x[1]])

# set ordering for convenience
to_order = ['Lin','MLP2','Att2K','Att2Q','MLP1', 'Att1K','Att1Q']
from_order = ['Tok','Att1V','MLP1','Att2VTok','Att2VAtt1','Att2VMLP1','MLP2','Lin']

df['from'] = pd.Categorical(df['from'], categories = from_order, ordered=True)
df['to'] = pd.Categorical(df['to'], categories=to_order, ordered=True)

df['logK'] = df['K'].apply(lambda x: np.log2(x))
# derivative wrt to t of 'kl_norm' grouping by seed, K, n, and ablated_edge
df['d_kl_norm'] = df.groupby(['seed', 'K', 'n', 'ablated_edge'])['kl_norm'].diff().fillna(0)
```

```python
def circuit_plot(ax, t, K, name: str, show_xlabels=False, derivative=False):
    t, filtered_df = filter_t(df, t)
    mask = (filtered_df['K'] == K)
    if derivative:
        snapshot = filtered_df[mask].pivot_table(values='d_kl_norm', index='from', columns='to', observed=False)
        v = np.max(np.abs(snapshot))
        sns.heatmap(snapshot, cmap='vlag', center=0, vmin=-v, vmax=v, square=True, ax = ax, cbar=True)
    else:
        snapshot = filtered_df[mask].pivot_table(values='kl_norm', index='from', columns='to', observed=False)
        sns.heatmap(snapshot, cmap='Grays', vmin=0, vmax = 1, square=True, ax = ax, cbar=True)
    ax.set_yticklabels(['Token', 'A1(Token)', 'MLP1', 'A2(Token)', 'A2(A1)', 'A2(MLP1)', 'MLP2'], rotation=0)
    ax.set_xticklabels(['Linear', 'MLP2', 'A2K', 'A2Q', 'MLP1', 'A1K', 'A1Q'], rotation=90)
    ax.tick_params(axis='both', which='both', length=0)
    ax.set_xlabel('')
    ax.set_ylabel('')
    ax.set_title(name)
    ax.invert_xaxis()
    # cbar = ax.collections[0].colorbar
    # cbar.set_label(r"$D_{KL}(\hat\pi | \tilde\pi)$")

    if not show_xlabels:
        ax.set_xticklabels([])
        ax.set_xlabel("")

    # draw boxes
    data = snapshot.to_numpy()
    rows, cols = np.where(~np.isnan(data))
    for r, c in zip(rows, cols):
        ax.add_patch(Rectangle(
            (c, r),       # (x, y) in data coordinates (col, row)
            1, 1,
            fill=False,
            edgecolor="gray",
            linewidth=1
        ))
```

```python
from __future__ import annotations

import math
import re
import xml.etree.ElementTree as ET
from pathlib import Path
from typing import Optional, Union

import numpy as np


_EDGE_ID_RE = re.compile(r"^e_(\d+)_(\d+)$")


def _strip_ns(tag: str) -> str:
    return tag.split("}", 1)[-1] if "}" in tag else tag


def _iter_elements(root: ET.Element):
    yield from root.iter()


def _patch_marker_units_strokewidth(root: ET.Element) -> int:
    """
    Force markerUnits="strokeWidth" on all <marker> elements so arrowheads
    (if implemented as markers) scale with stroke-width.
    Returns number of markers patched.
    """
    patched = 0
    for el in _iter_elements(root):
        if _strip_ns(el.tag) == "marker":
            if el.get("markerUnits") != "strokeWidth":
                el.set("markerUnits", "strokeWidth")
                patched += 1
    return patched


def _map_linear(x: float, lo: float, hi: float, out_lo: float, out_hi: float) -> float:
    if hi <= lo:
        return 0.5 * (out_lo + out_hi)
    t = (x - lo) / (hi - lo)
    t = max(0.0, min(1.0, t))
    return out_lo + t * (out_hi - out_lo)


def scale_svg_edges_from_W(
    svg_in: Union[str, Path],
    W: np.ndarray,
    svg_out: Optional[Union[str, Path]] = None,
    *,
    stroke_min: float = 0.75,
    stroke_max: float = 6.0,
    threshold: float = 0.0,
    log_scale: bool = False,
    set_opacity: bool = False,
    opacity_min: float = 0.15,
    opacity_max: float = 1.0,
    patch_markers: bool = True,
) -> str:
    """
    Modify an SVG (exported from Illustrator) by scaling each directed edge's stroke-width
    according to a weight matrix W.

    Conventions / assumptions
    - Each edge element has id="e_i_j" with integer i, j.
    - Directed edges correspond to the upper triangle of W (i < j).
    - If arrowheads are SVG markers, setting markerUnits="strokeWidth" makes them scale
      with stroke-width. (If Illustrator exported arrowheads as filled shapes, this will
      not rescale them.)

    Parameters
    ----------
    svg_in:
        Path to input SVG.
    W:
        Square numpy array of weights. Only entries with i<j are used.
    svg_out:
        If provided, write modified SVG to this path and also return the SVG text.
        If None, do not write; only return SVG text.
    stroke_min, stroke_max:
        Output stroke-width range (SVG user units).
    threshold:
        Weights <= threshold are hidden (stroke-width=0 and stroke-opacity=0).
        Note: if set_opacity=False, we still set stroke-opacity=0 for hidden edges.
    log_scale:
        If True, map log1p(weight) linearly.
    set_opacity:
        If True, set stroke-opacity based on weights (in addition to stroke-width).
    opacity_min, opacity_max:
        Output opacity range used when set_opacity=True.
    patch_markers:
        If True, set markerUnits="strokeWidth" for all SVG <marker> elements.

    Returns
    -------
    svg_text:
        The modified SVG as a UTF-8 string.

    Raises
    ------
    ValueError:
        If W is not square or if no upper-triangle weights exceed threshold.
    """
    W = np.asarray(W)
    if W.ndim != 2 or W.shape[0] != W.shape[1]:
        raise ValueError(f"W must be a square 2D array; got shape {W.shape}")

    n = W.shape[0]
    iu = np.triu_indices(n, k=1)
    upper_vals = W[iu].astype(float)

    active = upper_vals[upper_vals > threshold]
    if active.size == 0:
        raise ValueError(f"No upper-triangle weights exceed threshold={threshold}.")

    # Range for scaling
    w_lo = float(active.min())
    w_hi = float(active.max())

    if log_scale:
        # map using log1p; preserve a safe lower bound at 0
        def tx(x: float) -> float:
            return math.log1p(x)

        w_lo_t = tx(w_lo) if w_lo > 0 else 0.0
        w_hi_t = tx(w_hi) if w_hi > 0 else 0.0
    else:
        def tx(x: float) -> float:
            return x

        w_lo_t, w_hi_t = w_lo, w_hi

    # Parse SVG
    svg_in = Path(svg_in)
    tree = ET.parse(svg_in)
    root = tree.getroot()

    patched_markers = _patch_marker_units_strokewidth(root) if patch_markers else 0

    changed = 0
    for el in _iter_elements(root):
        el_id = el.get("id")
        if not el_id:
            continue
        m = _EDGE_ID_RE.match(el_id)
        if not m:
            continue

        i = int(m.group(1))
        j = int(m.group(2))
        if not (0 <= i < n and 0 <= j < n and i < j):
            # Ignore edges not matching your stated convention
            continue

        w = float(W[i, j])

        if w <= threshold:
            el.set("stroke-width", "0")
            el.set("stroke-opacity", "0")
            changed += 1
            continue

        w_t = tx(w)
        width = _map_linear(w_t, w_lo_t, w_hi_t, stroke_min, stroke_max)
        el.set("stroke-width", f"{width:.6g}")

        if set_opacity:
            op = _map_linear(w_t, w_lo_t, w_hi_t, opacity_min, opacity_max)
            el.set("stroke-opacity", f"{op:.6g}")

        changed += 1

    svg_bytes = ET.tostring(root, encoding="utf-8", xml_declaration=True)
    svg_text = svg_bytes.decode("utf-8")

    if svg_out is not None:
        svg_out = Path(svg_out)
        svg_out.parent.mkdir(parents=True, exist_ok=True)
        svg_out.write_text(svg_text, encoding="utf-8")

    # Return SVG text; in a notebook you can write it or display it.
    # Metadata for debugging (optional): you can inspect these by re-running with a wrapper.
    return svg_text
```

```python
import numpy as np
from graphviz import Digraph

def make_circuit_graph(W: np.ndarray, name: str, threshold = 0.1) -> Digraph:
    sources = [
        "Token",
        "A1(Token)",
        "MLP1",
        "A2(Token)",
        "A2(A1)",
        "A2(MLP1)",
        "MLP2",
    ]

    targets = [
        "Linear",
        "MLP2",
        "A2K",
        "A2Q",
        "MLP1",
        "A1K",
        "A1Q",
    ]

    all_nodes = sorted(set(sources + targets))

    outer_modules = {
        "Token":  ["Token"],
        "A1":     ["A1(Token)", "A1K", "A1Q"],
        "MLP1":   ["MLP1"],
        "A2":     ["A2(Token)", "A2(A1)", "A2(MLP1)", "A2K", "A2Q"],
        "MLP2":   ["MLP2"],
        "Linear": ["Linear"],
    }

    inner_modules = {
        "A1": ["A1K", "A1Q"],
        "A2": ["A2K", "A2Q"],
    }

    module_outputs = {
        "Token":  ["Token"],
        "A1":     ["A1(Token)"],
        "MLP1":   ["MLP1"],
        "A2":     ["A2(Token)", "A2(A1)", "A2(MLP1)"],
        "MLP2":   ["MLP2"],
        "Linear": ["Linear"],
    }

    module_order = {
        "Token":  ["Token"],
        "A1":     ["A1(Token)", "A1K", "A1Q"],
        "MLP1":   ["MLP1"],
        "A2":     ["A2(Token)", "A2(A1)", "A2(MLP1)", "A2K", "A2Q"],
        "MLP2":   ["MLP2"],
        "Linear": ["Linear"],
    }

    centered_outputs = [
        "Token",
        "A1(Token)",
        "MLP1",
        "A2(Token)",
        "MLP2",
        "Linear",
    ]

    assert W.shape == (len(sources), len(targets)), \
        f"W must be shape {(len(sources), len(targets))}, got {W.shape}"

    dot = Digraph("circuit", engine="dot")
    dot.attr(rankdir="TB", splines="true", concentrate="false", newrank="true")
    dot.attr("graph", fontsize="10", nodesep="0.30", ranksep="0.1")
    dot.attr("node", shape="ellipse", fontsize="8", width="0.0", height="0.0", fixedsize="false")
    dot.attr("edge", arrowsize="0.7")

    def nid(name_: str) -> str:
        return "n_" + str(abs(hash(name_)))

    id_of = {n: nid(n) for n in all_nodes}

    output_nodes = {n for vs in module_outputs.values() for n in vs}
    internal_nodes = {n for vs in inner_modules.values() for n in vs}

    def node_attrs(n: str) -> dict:
        if n in output_nodes:
            return dict(shape="box", penwidth="2.2")
        if n in internal_nodes:
            return dict(shape="ellipse", penwidth="1.0")
        return dict(shape="ellipse", penwidth="1.2")

    # ---------------------- FIXED COLUMNS (x) ----------------------
    col_order = ["Token", "A1", "MLP1", "A2", "MLP2", "Linear"]
    anchor_id = {m: f"ANCHOR_{m}" for m in col_order}

    # bind only ONE representative node per module to the column anchor
    rep_node = {}
    for m in col_order:
        outs = module_outputs.get(m, [])
        rep_node[m] = outs[0] if len(outs) else module_order[m][0]

    def bind_rep_to_column(mod: str):
        dot.edge(
            anchor_id[mod],
            id_of[rep_node[mod]],
            style="invis",
            weight="300000",
            constraint="false",  # do not affect vertical ranks
        )
    # ---------------------------------------------------------------

    # Outer clusters with nested K/Q clusters
    for mod in col_order:
        nodes = outer_modules[mod]
        with dot.subgraph(name=f"cluster_{mod}") as c:
            # cluster padding control; lower tightens the rounded box
            c.attr(label=mod, penwidth="1.6", style="rounded", margin="6")

            # invisible anchor INSIDE cluster (prevents left-floating dots)
            # c.node(
            #     anchor_id[mod],
            #     label="",
            #     shape="none",
            #     width="0",
            #     height="0",
            #     margin="0",
            #     style="invis",
            #     group=mod,
            # )

            if mod in inner_modules:
                with c.subgraph(name=f"cluster_{mod}_internal") as ci:
                    ci.attr(label="", penwidth="1.0", style="rounded,dashed", margin="3")
                    for n in inner_modules[mod]:
                        ci.node(id_of[n], label=n, group=mod, **node_attrs(n))

                for n in nodes:
                    if n in inner_modules[mod]:
                        continue
                    c.node(id_of[n], label=n, group=mod, **node_attrs(n))
            else:
                for n in nodes:
                    c.node(id_of[n], label=n, group=mod, **node_attrs(n))

            # vertical stacking inside module (centers content around representative)
            ordered = module_order.get(mod, nodes)
            for a, b in zip(ordered, ordered[1:]):
                c.edge(
                    id_of[a], id_of[b],
                    style="invis",
                    weight="200",
                    minlen="1",
                    constraint="true",
                )

    # Chain anchors left->right (global x ordering; edge-invariant)
    with dot.subgraph(name="anchor_row") as ar:
        ar.attr(rank="same")
        ar.attr("node",
            label="",
            shape="point",
            width="0.01",
            height="0.01",
            fixedsize="true",
            style="invis")
        for m in col_order:
            ar.node(anchor_id[m])
        for L, R in zip(col_order, col_order[1:]):
            ar.edge(
                anchor_id[L], anchor_id[R],
                style="invis",
                weight="600000",
                constraint="true",
            )

    # Bind each module's representative node to its anchor (ONE edge per module)
    for m in col_order:
        bind_rep_to_column(m)

    # Optional: align module outputs in the same rank (does not affect x ordering)
    with dot.subgraph(name="center_row") as r:
        r.attr(rank="same")
        for n in centered_outputs:
            r.node(id_of[n])

    # Spine for vertical progression through modules
    spine = ["Token", "A1(Token)", "MLP1", "A2(Token)", "MLP2", "Linear"]
    for a, b in zip(spine, spine[1:]):
        dot.edge(
            id_of[a], id_of[b],
            style="invis",
            weight="10000",
            minlen="1",
            constraint="true",
        )

    # Weighted edges (must not affect rank / ordering)
    thr = threshold
    absmax = float(np.max(np.abs(W))) if W.size else 0.0
    absmax = absmax if absmax > 0 else 1.0

    def rgba_hex(r, g, b, a):
        return "#{:02X}{:02X}{:02X}{:02X}".format(
            int(255 * r), int(255 * g), int(255 * b), int(255 * a)
        )

    for i, s in enumerate(sources):
        for j, tname in enumerate(targets):
            val = float(W[i, j])
            if abs(val) <= thr:
                continue

            mag = abs(val) / absmax
            alpha = 0.10 + 0.90 * mag
            penwidth = 0.6 + 3.0 * mag
            col = rgba_hex(1.0, 0.0, 0.0, alpha) if val < 0 else rgba_hex(0.0, 0.0, 1.0, alpha)

            dot.edge(
                id_of[s],
                id_of[tname],
                color=col,
                penwidth=str(penwidth),
                constraint="false",
                weight="1",
            )

    dot.render(name, format="svg", cleanup=True)
    return dot
```

```python
import numpy as np
from graphviz import Digraph

def make_circuit_graph(W: np.ndarray, name: str, threshold=0.1) -> Digraph:
    sources = [
        "Token",
        "A1(Token)",
        "MLP1",
        "A2(Token)",
        "A2(A1)",
        "A2(MLP1)",
        "MLP2",
    ]

    targets = [
        "Linear",
        "MLP2",
        "A2K",
        "A2Q",
        "MLP1",
        "A1K",
        "A1Q",
    ]

    all_nodes = sorted(set(sources + targets))

    outer_modules = {
        "Token":  ["Token"],
        "A1":     ["A1(Token)", "A1K", "A1Q"],
        "MLP1":   ["MLP1"],
        "A2":     ["A2(Token)", "A2(A1)", "A2(MLP1)", "A2K", "A2Q"],
        "MLP2":   ["MLP2"],
        "Linear": ["Linear"],
    }

    inner_modules = {
        "A1": ["A1K", "A1Q"],
        "A2": ["A2K", "A2Q"],
    }

    module_outputs = {
        "Token":  ["Token"],
        "A1":     ["A1(Token)"],
        "MLP1":   ["MLP1"],
        "A2":     ["A2(Token)", "A2(A1)", "A2(MLP1)"],
        "MLP2":   ["MLP2"],
        "Linear": ["Linear"],
    }

    module_order = {
        "Token":  ["Token"],
        "A1":     ["A1(Token)", "A1K", "A1Q"],
        "MLP1":   ["MLP1"],
        "A2":     ["A2(Token)", "A2(A1)", "A2(MLP1)", "A2K", "A2Q"],
        "MLP2":   ["MLP2"],
        "Linear": ["Linear"],
    }

    assert W.shape == (len(sources), len(targets)), \
        f"W must be shape {(len(sources), len(targets))}, got {W.shape}"

    dot = Digraph("circuit", engine="dot")
    dot.attr(rankdir="TB", splines="true", concentrate="false", newrank="true")
    dot.attr("graph", fontsize="10", nodesep="0.30", ranksep="0.1", margin="0")
    dot.attr("node", shape="ellipse", fontsize="8", fixedsize="false")
    dot.attr("edge", arrowsize="0.7")

    def nid(name_: str) -> str:
        return "n_" + str(abs(hash(name_)))

    id_of = {n: nid(n) for n in all_nodes}

    col_order = ["Token", "A1", "MLP1", "A2", "MLP2", "Linear"]

    # Representative node per module
    rep_node = {}
    for m in col_order:
        outs = module_outputs.get(m, [])
        rep_node[m] = outs[0] if len(outs) else module_order[m][0]

    # Deterministic left->right ordering key for representative nodes
    rep_sortv = {rep_node[m]: str(i) for i, m in enumerate(col_order)}

    output_nodes = {n for vs in module_outputs.values() for n in vs}
    internal_nodes = {n for vs in inner_modules.values() for n in vs}

    def node_attrs(n: str) -> dict:
        if n in output_nodes:
            return dict(shape="box", penwidth="2.2")
        if n in internal_nodes:
            return dict(shape="ellipse", penwidth="1.0")
        return dict(shape="ellipse", penwidth="1.2")

    def maybe_sortv(n: str) -> dict:
        # Only reps get sortv (they're the ones in the single horizontal rank)
        return {"sortv": rep_sortv[n]} if n in rep_sortv else {}

    cluster_modules = {"A1", "A2"}

    # Emit nodes / clusters
    for mod in col_order:
        nodes = outer_modules[mod]

        if mod not in cluster_modules:
            assert len(nodes) == 1, (mod, nodes)
            n = nodes[0]
            dot.node(id_of[n], label=n, group=mod, **maybe_sortv(n), **node_attrs(n))
            continue

        with dot.subgraph(name=f"cluster_{mod}") as c:
            c.attr(label=mod, penwidth="1.6", style="rounded", margin="5")

            if mod in inner_modules:
                with c.subgraph(name=f"cluster_{mod}_internal") as ci:
                    ci.attr(label="", penwidth="1.0", style="rounded,dashed", margin="2")
                    for n in inner_modules[mod]:
                        ci.node(id_of[n], label=n, group=mod, **maybe_sortv(n), **node_attrs(n))

                for n in nodes:
                    if n in inner_modules[mod]:
                        continue
                    c.node(id_of[n], label=n, group=mod, **maybe_sortv(n), **node_attrs(n))
            else:
                for n in nodes:
                    c.node(id_of[n], label=n, group=mod, **maybe_sortv(n), **node_attrs(n))

            # vertical stacking INSIDE module only
            ordered = module_order.get(mod, nodes)
            for a, b in zip(ordered, ordered[1:]):
                c.edge(id_of[a], id_of[b], style="invis", weight="200", minlen="1", constraint="true")

    # One horizontal row containing only the representatives
    with dot.subgraph(name="modules_row") as mr:
        mr.attr(rank="same")
        for m in col_order:
            mr.node(id_of[rep_node[m]])

    # Weighted edges (do not affect rank)
    thr = threshold
    absmax = float(np.max(np.abs(W))) if W.size else 0.0
    absmax = absmax if absmax > 0 else 1.0

    def rgba_hex(r, g, b, a):
        return "#{:02X}{:02X}{:02X}{:02X}".format(
            int(255 * r), int(255 * g), int(255 * b), int(255 * a)
        )

    for i, s in enumerate(sources):
        for j, tname in enumerate(targets):
            val = float(W[i, j])
            if abs(val) <= thr:
                continue

            mag = abs(val) / absmax
            alpha = 0.10 + 0.90 * mag
            penwidth = 0.6 + 3.0 * mag
            col = rgba_hex(1.0, 0.0, 0.0, alpha) if val < 0 else rgba_hex(0.0, 0.0, 1.0, alpha)

            dot.edge(id_of[s], id_of[tname],
                     color=col, penwidth=str(penwidth),
                     constraint="false", weight="1")

    dot.render(name, format="svg", cleanup=True)
    return dot
```

```python
import numpy as np
from graphviz import Digraph
import hashlib
from typing import Dict, Tuple

def _stable_id(s: str) -> str:
    # Avoid Python's randomized hash(); Graphviz node IDs must be stable across runs.
    return "n_" + hashlib.md5(s.encode("utf-8")).hexdigest()

def _parse_plain_positions(plain: str) -> Dict[str, Tuple[float, float]]:
    """
    Parse Graphviz 'plain' output. Returns {node_id: (x, y)}.
    Coordinates are in inches in plain format.
    """
    pos = {}
    for line in plain.splitlines():
        if not line:
            continue
        parts = line.split()
        if parts[0] == "node":
            # format: node <name> <x> <y> <width> <height> <label> ...
            name = parts[1]
            x = float(parts[2])
            y = float(parts[3])
            pos[name] = (x, y)
    return pos

def make_circuit_graph(W: np.ndarray, name: str, threshold=0.1) -> Digraph:
    sources = [
        "Token", "A1(Token)", "MLP1", "A2(Token)", "A2(A1)", "A2(MLP1)", "MLP2",
    ]
    targets = ["Linear", "MLP2", "A2K", "A2Q", "MLP1", "A1K", "A1Q"]

    assert W.shape == (len(sources), len(targets)), \
        f"W must be shape {(len(sources), len(targets))}, got {W.shape}"

    all_nodes = sorted(set(sources + targets))
    id_of = {n: _stable_id(n) for n in all_nodes}

    outer_modules = {
        "Token":  ["Token"],
        "A1":     ["A1(Token)", "A1K", "A1Q"],
        "MLP1":   ["MLP1"],
        "A2":     ["A2(Token)", "A2(A1)", "A2(MLP1)", "A2K", "A2Q"],
        "MLP2":   ["MLP2"],
        "Linear": ["Linear"],
    }
    inner_modules = {"A1": ["A1K", "A1Q"], "A2": ["A2K", "A2Q"]}
    module_outputs = {
        "Token":  ["Token"],
        "A1":     ["A1(Token)"],
        "MLP1":   ["MLP1"],
        "A2":     ["A2(Token)", "A2(A1)", "A2(MLP1)"],
        "MLP2":   ["MLP2"],
        "Linear": ["Linear"],
    }
    module_order = {
        "Token":  ["Token"],
        "A1":     ["A1(Token)", "A1K", "A1Q"],
        "MLP1":   ["MLP1"],
        "A2":     ["A2(Token)", "A2(A1)", "A2(MLP1)", "A2K", "A2Q"],
        "MLP2":   ["MLP2"],
        "Linear": ["Linear"],
    }

    col_order = ["Token", "A1", "MLP1", "A2", "MLP2", "Linear"]
    rep_node = {m: (module_outputs[m][0] if module_outputs.get(m) else module_order[m][0]) for m in col_order}

    output_nodes = {n for vs in module_outputs.values() for n in vs}
    internal_nodes = {n for vs in inner_modules.values() for n in vs}

    def node_attrs(n: str) -> dict:
        if n in output_nodes:
            return dict(shape="box", penwidth="2.2")
        if n in internal_nodes:
            return dict(shape="ellipse", penwidth="1.0")
        return dict(shape="ellipse", penwidth="1.2")

    cluster_modules = {"A1", "A2"}

    def build_graph(include_W_edges: bool, engine: str) -> Digraph:
        g = Digraph("circuit", engine=engine)
        # NOTE: For pinned pass we will use neato -n2; for baseline pass use dot.
        g.attr(rankdir="TB", splines="true", concentrate="false", newrank="true")
        g.attr("graph", fontsize="10", nodesep="0.30", ranksep="0.1", margin="0")
        g.attr("node", shape="ellipse", fontsize="8", fixedsize="false")
        g.attr("edge", arrowsize="0.7")

        # --- emit nodes/clusters ---
        for mod in col_order:
            nodes = outer_modules[mod]
            if mod not in cluster_modules:
                assert len(nodes) == 1, (mod, nodes)
                n = nodes[0]
                g.node(id_of[n], label=n, group=mod, **node_attrs(n))
                continue

            with g.subgraph(name=f"cluster_{mod}") as c:
                c.attr(label=mod, penwidth="1.6", style="rounded", margin="5")

                if mod in inner_modules:
                    with c.subgraph(name=f"cluster_{mod}_internal") as ci:
                        ci.attr(label="", penwidth="1.0", style="rounded,dashed", margin="2")
                        for n in inner_modules[mod]:
                            ci.node(id_of[n], label=n, group=mod, **node_attrs(n))

                    for n in nodes:
                        if n in inner_modules[mod]:
                            continue
                        c.node(id_of[n], label=n, group=mod, **node_attrs(n))
                else:
                    for n in nodes:
                        c.node(id_of[n], label=n, group=mod, **node_attrs(n))

                # vertical stacking INSIDE module
                ordered = module_order.get(mod, nodes)
                for a, b in zip(ordered, ordered[1:]):
                    c.edge(id_of[a], id_of[b], style="invis", weight="200", minlen="1", constraint="true")

        # --- anchors: fix L->R order independent of W ---
        anchor_id = {m: f"ANCHOR_{m}" for m in col_order}
        with g.subgraph(name="anchor_row") as ar:
            ar.attr(rank="same")
            ar.attr("node", label="", shape="point", width="0.01", height="0.01",
                    fixedsize="true", style="invis")
            for m in col_order:
                ar.node(anchor_id[m], group=m)
            for L, R in zip(col_order, col_order[1:]):
                ar.edge(anchor_id[L], anchor_id[R], style="invis", constraint="true", weight="10000", minlen="1")

        for m in col_order:
            g.edge(anchor_id[m], id_of[rep_node[m]], style="invis", constraint="false", weight="10000", minlen="0")

        # --- force reps onto one horizontal row ---
        with g.subgraph(name="modules_row") as mr:
            mr.attr(rank="same")
            for m in col_order:
                mr.node(id_of[rep_node[m]])

        # --- W edges (optional) ---
        if include_W_edges:
            thr = threshold
            absmax = float(np.max(np.abs(W))) if W.size else 0.0
            absmax = absmax if absmax > 0 else 1.0

            def rgba_hex(r, g_, b, a):
                return "#{:02X}{:02X}{:02X}{:02X}".format(
                    int(255 * r), int(255 * g_), int(255 * b), int(255 * a)
                )

            for i, s in enumerate(sources):
                for j, tname in enumerate(targets):
                    val = float(W[i, j])
                    if abs(val) <= thr:
                        continue
                    mag = abs(val) / absmax
                    alpha = 0.10 + 0.90 * mag
                    penwidth = 0.6 + 3.0 * mag
                    col = rgba_hex(1.0, 0.0, 0.0, alpha) if val < 0 else rgba_hex(0.0, 0.0, 1.0, alpha)
                    g.edge(id_of[s], id_of[tname], color=col, penwidth=str(penwidth),
                           constraint="false", weight="1")
        return g

    # PASS 1: baseline layout without W-edges using dot
    base = build_graph(include_W_edges=False, engine="dot")
    plain = base.pipe(format="plain").decode("utf-8")
    pos = _parse_plain_positions(plain)

    # PASS 2: final render with pinned node positions using neato -n2
    final = build_graph(include_W_edges=True, engine="neato")
    final.attr("graph", overlap="false")  # optional

    # pin ALL nodes that have positions in the baseline layout (includes ANCHOR_*)
    for node_id, (x, y) in pos.items():
        final.node(node_id, pos=f"{x},{y}!")

    # IMPORTANT: neato_no_op=2 tells neato to use existing positions (no re-layout)
    final.render(name, format="svg", cleanup=True, neato_no_op=2)
    return final
```

```python
import numpy as np
from graphviz import Digraph
import hashlib
from typing import Dict, Tuple

def _stable_id(s: str) -> str:
    return "n_" + hashlib.md5(s.encode("utf-8")).hexdigest()

def _parse_plain_positions(plain: str) -> Dict[str, Tuple[float, float]]:
    pos = {}
    for line in plain.splitlines():
        parts = line.split()
        if parts and parts[0] == "node":
            # node <id> <x> <y> ...
            pos[parts[1]] = (float(parts[2]), float(parts[3]))  # inches
    return pos

def _rgba_hex(r, g, b, a):
    return "#{:02X}{:02X}{:02X}{:02X}".format(
        int(255 * r), int(255 * g), int(255 * b), int(255 * a)
    )

def make_circuit_graph(W: np.ndarray, name: str, threshold=0.1) -> Digraph:
    sources = ["Token","A1(Token)","MLP1","A2(Token)","A2(A1)","A2(MLP1)","MLP2"]
    targets = ["Linear","MLP2","A2K","A2Q","MLP1","A1K","A1Q"]
    assert W.shape == (len(sources), len(targets))

    all_nodes = sorted(set(sources + targets))
    id_of = {n: _stable_id(n) for n in all_nodes}

    # ----- PASS 1: baseline layout with DOT (no W edges) -----
    base = Digraph("circuit", engine="dot")
    base.attr(rankdir="TB", splines="true", concentrate="false", newrank="true")
    base.attr("graph", fontsize="10", nodesep="0.30", ranksep="0.1", margin="0")
    base.attr("node", shape="ellipse", fontsize="8", fixedsize="false")

    # put nodes (baseline-only: just nodes; keep your baseline scaffolding here if desired)
    for n in all_nodes:
        base.node(id_of[n], label=n)

    # If you want a fixed baseline order, enforce it here (baseline-only)
    col_order = ["Token", "A1(Token)", "MLP1", "A2(Token)", "MLP2", "Linear"]
    with base.subgraph(name="baseline_row") as r:
        r.attr(rank="same")
        for n in col_order:
            r.node(id_of[n])
        for a, b in zip(col_order, col_order[1:]):
            r.edge(id_of[a], id_of[b], style="invis", constraint="true", weight="10000")

    plain = base.pipe(format="plain").decode("utf-8")
    pos_in = _parse_plain_positions(plain)

    # ----- PASS 2: pinned draw with NEATO -n2 (nodes+W edges only) -----
    final = Digraph("circuit", engine="neato")
    final.attr("graph", overlap="false", splines="true", margin="0")
    final.attr("node", shape="ellipse", fontsize="8", fixedsize="false")
    final.attr("edge", arrowsize="0.7")

    # pin all nodes; convert inches -> points (×72)
    for n in all_nodes:
        nid = id_of[n]
        x_in, y_in = pos_in[nid]
        x_pt, y_pt = 72.0 * x_in, 72.0 * y_in
        final.node(nid, label=n, pos=f"{x_pt},{y_pt}!")

    # add W edges (won't affect node positions under -n2)
    thr = threshold
    absmax = float(np.max(np.abs(W))) if W.size else 1.0
    absmax = absmax if absmax > 0 else 1.0

    for i, s in enumerate(sources):
        for j, t in enumerate(targets):
            val = float(W[i, j])
            if abs(val) <= thr:
                continue
            mag = abs(val) / absmax
            alpha = 0.10 + 0.90 * mag
            penwidth = 0.6 + 3.0 * mag
            col = _rgba_hex(1.0, 0.0, 0.0, alpha) if val < 0 else _rgba_hex(0.0, 0.0, 1.0, alpha)
            final.edge(id_of[s], id_of[t], color=col, penwidth=str(penwidth))

    final.render(name, format="svg", cleanup=True, neato_no_op=2)
    return final
```

```python
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

def tsne(data, labels, title, ax):
    # Convert to numpy
    x_np = data.detach().cpu().numpy()
    labels_np = labels.detach().cpu().numpy()

    # Run t-SNE
    tsne = TSNE(
        n_components=2,
        perplexity=30.0,
        learning_rate="auto",
        init="pca",
        random_state=0,
    )
    x_2d = tsne.fit_transform(x_np)   # (B, 2)

    # Build dataframe for seaborn
    df = pd.DataFrame({
        "tsne_1": x_2d[:, 0],
        "tsne_2": x_2d[:, 1],
        "label": labels_np.astype(int)
    })

    sns.scatterplot(
        data=df,
        x="tsne_1",
        y="tsne_2",
        hue="label",
        palette="tab20",
        s=3,
        alpha=0.8,
        edgecolor=None,
        ax = ax
    )

    ax.text(0.05, 0.05, title, ha='left', va='bottom', transform=ax.transAxes)
    ax.legend().remove()
    ax.set_xlabel('')
    ax.set_ylabel('')
    ax.set_xticks([])
    ax.set_yticks([])
```

```python
def tv_tsne(param_idx, seed_idx, n_idx, t, name, ax, B = 2**10):
    tv, tasks, _ = get_tv(param_idx, seed_idx, n_idx, t)
    data = tv[2][:, -1, :]
    tsne(data, tasks, name, ax)
```

```python
# fix settings
param_idx, seed_idx, n_idx, t = 0, 0, 0, 1e5

tv, tasks, task_model = get_tv(param_idx, seed_idx, n_idx, t)
tv = tv[2][:, -1, :]
tv = tv / tv.norm(dim=-1, keepdim=True)
tv = tv.detach().cpu().numpy()
tasks = tasks.detach().cpu().numpy()
task_pool = task_model.task_pool.float()
stat_dist = task_model.stat_dist.float()
bigram_probs = task_pool * stat_dist.unsqueeze(-1)
bigram_probs = bigram_probs.flatten(-2, -1).cpu().numpy()
```

```python
# compute pairwise symmetric kl divergence between bigram_probs for first dimension
from scipy.special import rel_entr
def kl_divergence(p, q):
    return 0.5*np.sum(rel_entr(p, q)) + 0.5*np.sum(rel_entr(q, p))

n_tasks = bigram_probs.shape[0]
kl_matrix = np.zeros((n_tasks, n_tasks))
for i in range(n_tasks):
    for j in range(n_tasks):
        kl_matrix[i, j] = kl_divergence(bigram_probs[i], bigram_probs[j])
```

```python
# get indices of unique tasks
unique_tasks, indices = np.unique(tasks, return_index=True)
# get corresponding tv vectors
unique_tv = tv[indices]
tv_cos_sim = unique_tv @ unique_tv.T
```

#### patching

```python
def accuracy(logits1, logits2):
    assert logits1.shape == logits2.shape
    n = logits1.shape[0]
    p1 = F.softmax(logits1, -1)
    p2 = F.softmax(logits2, -1)
    _, idx1 = p1.max(-1)
    _, idx2 = p2.max(-1)

    return (torch.sum(idx1 == idx2)/n).item()

def similarity(logits1, logits2):
    assert logits1.shape == logits2.shape
    
    return math.exp(-KL(logits1, logits2)/KL(logits1, torch.zeros_like(logits1)))
    # return 1 - KL(logits1, logits2)/KL(logits1, torch.zeros_like(logits1))
```

```python
from pd.compute import naming
from base.utils import KL
from base.estimators.markov_model_ar import BiMem, UniMem
from base.operations.toolkit import get_res_act, patched_res_act_performance
from matplotlib import patches as mpatches

def perturb_plot(ax, seed_idx, n_idx, param_idx, t, name: str, estimator: str, attn_idx: int, minimal = False):
    K = exp_config.paramrange[param_idx]

    get_model_safe = u.dynamic_import(exp_config.model, 'get_model_safe')
    get_task_model_safe = u.dynamic_import(exp_config.task_model, 'get_task_model_safe')
    model_name = naming(exp_config.seedrange[seed_idx], exp_config.nrange[n_idx], exp_config.param_name, K)

    config = u.load_config_and_task_pool(Path(exp_config.base_dir) / 'data' / model_name)
    model = get_model_safe(config)
    task_model = get_task_model_safe(config)
    model.eval()
    state_idx = get_state_idx(config, t)
    state_dict = torch.load(Path(exp_config.base_dir) / 'data' / model_name / 'state' / f'{state_idx}.pt')
    s = state_dict['iter']
    state_dict = state_dict['state']
    model.load_state_dict(state_dict)
    B = 2**10
    N = exp_config.nrange[n_idx]
    xa, ya, tasks = task_model.get_batch(B, config.training.context_len, return_tasks=True)
    xb, yb = task_model.get_batch(B, config.training.context_len)
    if estimator == '2_mem':
        estimator = BiMem(config, task_model.task_pool.float(), task_model.stat_dist.float())
    elif estimator == '1_mem':
        estimator = UniMem(config, task_model.task_pool.float(), task_model.stat_dist.float())
    
    # (x, y) = posterior on x, predict on current states of y
    behavior = {}
    behavior['AA'], _, hatTa = estimator(xa, ya, return_hatT=True)
    behavior['BB'], _, hatTb = estimator(xb, yb, return_hatT=True)
    behavior['AB'] = torch.log(torch.einsum('ijkl,ijk->ijl', hatTa, F.one_hot(xb, config.task.vocab_size).float()))
    behavior['BA'] = torch.log(torch.einsum('ijkl,ijk->ijl', hatTb, F.one_hot(xa, config.task.vocab_size).float()))

    a_logits, _ = model(xa, ya)
    b_logits, _ = model(xb, yb)
    a_tv = get_res_act(model, (xa, ya))
    b_tv = get_res_act(model, (xb, yb))

    output = {
        'predictor': [],
        'accuracy': [],
        'condition': [],
        'similarity': [],
    }
    
    bpa_logits, _ = patched_res_act_performance(model, [attn_idx], [a_tv[attn_idx]], (xb, yb))
    apb_logits, _ = patched_res_act_performance(model, [attn_idx], [b_tv[attn_idx]], (xa, ya))
    for key in behavior.keys():
        t = behavior[key]

        output['similarity'].append(similarity(t.flatten(0,1), a_logits))
        output['accuracy'].append(accuracy(t.flatten(0,1), a_logits))
        output['predictor'].append(key)
        output['condition'].append('SeqA')

        output['similarity'].append(similarity(t.flatten(0,1), b_logits))
        output['accuracy'].append(accuracy(t.flatten(0,1), b_logits))
        output['predictor'].append(key)
        output['condition'].append('SeqB')

        output['similarity'].append(similarity(t.flatten(0,1), bpa_logits))
        output['accuracy'].append(accuracy(t.flatten(0,1), bpa_logits))
        output['predictor'].append(key)
        output['condition'].append('SeqB (Patch A)')

        output['similarity'].append(similarity(t.flatten(0,1), apb_logits))
        output['accuracy'].append(accuracy(t.flatten(0,1), apb_logits))
        output['predictor'].append(key)
        output['condition'].append('SeqA (Patch B)')
    
    output = pd.DataFrame(output)

    baseline_similarity = similarity(t.flatten(0,1), torch.zeros_like(apb_logits))  # unconditioned baseline

    if minimal:
        plot_data = output[output['condition'].isin(['SeqA', 'SeqB (Patch A)']) & output['predictor'].isin(['AA', 'AB'])].copy()
    else:
        plot_data = output.copy()

    if minimal:
        hue_order = ['SeqA', 'SeqB (Patch A)']
        hatches = ['', '']
    else:
        hue_order = ['SeqA', 'SeqB', 'SeqB (Patch A)', 'SeqA (Patch B)']
        hatches = ['/', '\\\\', 'xx', '..']

    sns.barplot(
        data=plot_data,
        x='predictor',
        y='similarity',
        hue='condition',
        hue_order=hue_order,
        ax=ax,
        legend=True,
        palette=['black', 'white'],
        edgecolor='black',
    )

    # for i, bar in enumerate(ax.patches):
    #     bar.set_facecolor('white' if i // 2 ==1 else 'black')
    #     bar.set_edgecolor('black')
    #     # bar.set_hatch(hatches[i % len(hue_order)])

    # legend_handles = [
    #     mpatches.Patch(facecolor='black', edgecolor='white', hatch=h, label=label)
    #     for h, label in zip(hatches, hue_order)
    # ]
    # ax.legend(handles=legend_handles, title='condition')
    # sns.barplot(data = plot_data, x = 'predictor', y = 'similarity', hue='condition', ax = ax, legend=True, palette=['C6', 'C9', 'C6', 'C9'])
    # ax.set_title(f'{name} Task Vector Patching')
    ax.set_xlabel('Predictor')
    ax.set_ylabel('Similarity')
    # ax.axhline(baseline_similarity, color='black', linestyle=':', label='Uniform Baseline')
    
    return output, behavior, ax
```

```python
from matplotlib.colors import LinearSegmentedColormap, to_rgba
from matplotlib import colormaps

def black_to_color(name=None, final_color=None, N=256, reverse=True):
    """
    Return a colormap that goes between black and a target color.

    Parameters
    ----------
    name : str, optional
        Name of a Matplotlib colormap (e.g. 'Reds'). Used if final_color is None.
    final_color : str or tuple, optional
        The final color (matplotlib-recognized format). If None, take from colormap.
    N : int
        Number of levels in the colormap.
    reverse : bool
        If True, make the gradient go from the target color to black.
    """
    if final_color is None:
        if name is None:
            raise ValueError("Either `name` or `final_color` must be provided.")
        base = colormaps.get_cmap(name)
        target = base(1.0)  # brightest color
    else:
        target = to_rgba(final_color)  # convert to RGBA

    colors = [(0, 0, 0, 1), target]
    if reverse:
        colors = colors[::-1]

    cmap = LinearSegmentedColormap.from_list(
        (name or "custom") + "_black" + ("_rev" if reverse else ""),
        colors,
        N=N
    )
    return cmap

def log_heatmap(pvt, ax=None, vmin=None, vmax=None, cmap="viridis", rasterized=True):
    if ax is None:
        ax = plt.gca()

    X = np.asarray(pvt.columns, dtype=float)
    Y = np.asarray(pvt.index, dtype=float)
    Z = pvt.values

    def log_edges(vals):
        vals = np.asarray(vals, float)
        ratios = vals[1:] / vals[:-1]
        edges = np.empty(len(vals) + 1)
        edges[1:-1] = np.sqrt(vals[:-1] * vals[1:])
        edges[0] = vals[0] / np.sqrt(ratios[0])
        edges[-1] = vals[-1] * np.sqrt(ratios[-1])
        return edges

    x_edges = log_edges(X)
    y_edges = log_edges(Y)

    qm = ax.pcolormesh(
        x_edges, y_edges, Z,
        shading="auto",
        vmin=vmin, vmax=vmax, cmap=cmap,
        edgecolors="none",
        linewidth=0,
        antialiased=False,
        rasterized=rasterized,   # <-- key
    )

    ax.set_xscale("log")
    ax.set_yscale("log", base=2)

    for spine in ax.spines.values():
        spine.set_visible(False)

    return ax, qm

import numpy as np
import matplotlib.pyplot as plt

def composite_cmap_colorbar(
    fig,
    colors,
    *,
    vmin,
    vmax,
    ax_rect=(1.05, 0.2, 0.025, 0.7),   # (x, y, w_per_cmap, h) in figure coords
    n=256,                             # vertical resolution
    ticks=None,
    tick_right=True,
    label="",
    remove_spines=True,
    separator=True,
    separator_kwargs=None,
    cmap_factory=None,                 # function(color)->cmap; defaults to black_to_color(final_color=color)
):
    """
    Draw a composite vertical "colorbar" consisting of adjacent columns, one per colormap,
    all sharing the same [vmin, vmax] normalization.

    Parameters
    ----------
    fig : matplotlib.figure.Figure
    colors : sequence
        Sequence of colors (any matplotlib-acceptable color objects).
    vmin, vmax : float
    ax_rect : (x, y, w_per_cmap, h)
        Rectangle in figure coordinates. Width is per-colormap; total width scales with len(colors).
    n : int
        Number of samples along the vertical axis.
    ticks : sequence or None
        Y ticks in data units. If None, no ticks are set.
    tick_right : bool
        Place y ticks on the right.
    label : str
        Optional y-axis label (acts like a colorbar label).
    remove_spines : bool
        Hide all spines.
    separator : bool
        Draw vertical separators between columns.
    separator_kwargs : dict or None
        Passed to ax.vlines; defaults to thin white separators.
    cmap_factory : callable or None
        If provided, called as cmap_factory(color) -> matplotlib Colormap.
        If None, uses black_to_color(final_color=color).

    Returns
    -------
    ax : matplotlib.axes.Axes
    """
    if cmap_factory is None:
        def cmap_factory(c):
            return black_to_color(final_color=c)

    x, y, w_per, h = ax_rect
    k = len(colors)
    if k == 0:
        raise ValueError("colors must be a non-empty sequence.")

    ax = fig.add_axes([x, y, w_per * k, h])

    values = np.linspace(vmin, vmax, n)[:, None]  # (n, 1)
    norm = plt.Normalize(vmin=vmin, vmax=vmax)

    for i, c in enumerate(colors):
        ax.imshow(
            values,
            cmap=cmap_factory(c),
            norm=norm,
            aspect="auto",
            origin="lower",
            extent=(i, i + 1, vmin, vmax),
        )

    ax.set_xlim(0, k)
    ax.set_ylim(vmin, vmax)
    ax.set_xticks([])

    if ticks is not None:
        ax.set_yticks(list(ticks))

    if tick_right:
        ax.yaxis.tick_right()
        ax.yaxis.set_label_position("right")

    if label:
        ax.set_ylabel(label)

    if remove_spines:
        for spine in ax.spines.values():
            spine.set_visible(False)

    if separator and k > 1:
        if separator_kwargs is None:
            separator_kwargs = dict(colors="white", linewidth=0.5)
        ax.vlines(np.arange(1, k), vmin, vmax, **separator_kwargs)

    return ax
```

### loss

```python
def plot_loss(cur_ax, model_perf, est_perf, exp_config,
                  est_idx_list: list = list(range(4)),
                  dist: list = ['train', 'val'],
                  alpha_val: float = 0.5,
                  plot_kwargs: dict = {}):

    param_name = exp_config.param_name

    # --- estimator lines first ---
    data = est_perf
    palette = sns.color_palette()

    legend_colors = [palette[1], palette[2], palette[3], palette[0]]
    legend_names = ['1-Mem', '1-Gen', '2-Mem', '2-Gen']
    color_dict = {name: color for name, color in zip(legend_names, legend_colors)}

    est_names = ['1_mem_last', '1_gen_last', '2_mem_last', '2_gen_last']

    legend_handles = []
    for i in est_idx_list:
        for d in dist:
            est = est_names[i]
            mask = (data['name'] == est) & (data['dist'] == d)
            color = color_dict[legend_names[i]]
            cur_ax.axhline(
                y=data[mask]['loss'].unique().item(),
                label=f'{est}',
                color=color,
                alpha=alpha_val if d == 'val' else 1.0,
                **plot_kwargs
            )

        legend_handles.append(
            mlines.Line2D([], [], color=color, linestyle='-', label=f'{est}')
        )

    # --- model curves last (on top) ---
    data = model_perf
    color = 'black'

    for d in dist:
        mask = data['dist'] == d
        sns.lineplot(
            data=data[mask],
            x='t',
            y='loss',
            label=d,
            ax=cur_ax,
            color=color,
            alpha=alpha_val if d == 'val' else 1.0,
            **plot_kwargs
        )
    
    legend_handles.append(
        mlines.Line2D([], [], color=color, linestyle='-', label='Model')
    )

    cur_ax.get_legend().remove()

    cur_ax.set_title(rf'${param_name}={data[param_name].unique().item()}$')
    cur_ax.set_xlabel(r'$t$')
    cur_ax.set_ylabel('Loss')

    return legend_handles
```

```python
from matplotlib.colors import LinearSegmentedColormap, to_rgba
from matplotlib import colormaps

def black_to_color(name=None, final_color=None, N=256, reverse=True):
    """
    Return a colormap that goes between black and a target color.

    Parameters
    ----------
    name : str, optional
        Name of a Matplotlib colormap (e.g. 'Reds'). Used if final_color is None.
    final_color : str or tuple, optional
        The final color (matplotlib-recognized format). If None, take from colormap.
    N : int
        Number of levels in the colormap.
    reverse : bool
        If True, make the gradient go from the target color to black.
    """
    if final_color is None:
        if name is None:
            raise ValueError("Either `name` or `final_color` must be provided.")
        base = colormaps.get_cmap(name)
        target = base(1.0)  # brightest color
    else:
        target = to_rgba(final_color)  # convert to RGBA

    colors = [(0, 0, 0, 1), target]
    if reverse:
        colors = colors[::-1]

    cmap = LinearSegmentedColormap.from_list(
        (name or "custom") + "_black" + ("_rev" if reverse else ""),
        colors,
        N=N
    )
    return cmap

def log_heatmap(pvt, ax=None, vmin=None, vmax=None, cmap="viridis", rasterized=True):
    if ax is None:
        ax = plt.gca()

    X = np.asarray(pvt.columns, dtype=float)
    Y = np.asarray(pvt.index, dtype=float)
    Z = pvt.values

    def log_edges(vals):
        vals = np.asarray(vals, float)
        ratios = vals[1:] / vals[:-1]
        edges = np.empty(len(vals) + 1)
        edges[1:-1] = np.sqrt(vals[:-1] * vals[1:])
        edges[0] = vals[0] / np.sqrt(ratios[0])
        edges[-1] = vals[-1] * np.sqrt(ratios[-1])
        return edges

    x_edges = log_edges(X)
    y_edges = log_edges(Y)

    qm = ax.pcolormesh(
        x_edges, y_edges, Z,
        shading="auto",
        vmin=vmin, vmax=vmax, cmap=cmap,
        edgecolors="none",
        linewidth=0,
        antialiased=False,
        rasterized=rasterized,   # <-- key
    )

    ax.set_xscale("log")
    ax.set_yscale("log", base=2)

    for spine in ax.spines.values():
        spine.set_visible(False)

    return ax, qm

import numpy as np
import matplotlib.pyplot as plt

def composite_cmap_colorbar(
    fig,
    colors,
    *,
    vmin,
    vmax,
    ax_rect=(1.05, 0.2, 0.025, 0.7),   # (x, y, w_per_cmap, h) in figure coords
    n=256,                             # vertical resolution
    ticks=None,
    tick_right=True,
    label="",
    remove_spines=True,
    separator=True,
    separator_kwargs=None,
    cmap_factory=None,                 # function(color)->cmap; defaults to black_to_color(final_color=color)
):
    """
    Draw a composite vertical "colorbar" consisting of adjacent columns, one per colormap,
    all sharing the same [vmin, vmax] normalization.

    Parameters
    ----------
    fig : matplotlib.figure.Figure
    colors : sequence
        Sequence of colors (any matplotlib-acceptable color objects).
    vmin, vmax : float
    ax_rect : (x, y, w_per_cmap, h)
        Rectangle in figure coordinates. Width is per-colormap; total width scales with len(colors).
    n : int
        Number of samples along the vertical axis.
    ticks : sequence or None
        Y ticks in data units. If None, no ticks are set.
    tick_right : bool
        Place y ticks on the right.
    label : str
        Optional y-axis label (acts like a colorbar label).
    remove_spines : bool
        Hide all spines.
    separator : bool
        Draw vertical separators between columns.
    separator_kwargs : dict or None
        Passed to ax.vlines; defaults to thin white separators.
    cmap_factory : callable or None
        If provided, called as cmap_factory(color) -> matplotlib Colormap.
        If None, uses black_to_color(final_color=color).

    Returns
    -------
    ax : matplotlib.axes.Axes
    """
    if cmap_factory is None:
        def cmap_factory(c):
            return black_to_color(final_color=c)

    x, y, w_per, h = ax_rect
    k = len(colors)
    if k == 0:
        raise ValueError("colors must be a non-empty sequence.")

    ax = fig.add_axes([x, y, w_per * k, h])

    values = np.linspace(vmin, vmax, n)[:, None]  # (n, 1)
    norm = plt.Normalize(vmin=vmin, vmax=vmax)

    for i, c in enumerate(colors):
        ax.imshow(
            values,
            cmap=cmap_factory(c),
            norm=norm,
            aspect="auto",
            origin="lower",
            extent=(i, i + 1, vmin, vmax),
        )

    ax.set_xlim(0, k)
    ax.set_ylim(vmin, vmax)
    ax.set_xticks([])

    if ticks is not None:
        ax.set_yticks(list(ticks))

    if tick_right:
        ax.yaxis.tick_right()
        ax.yaxis.set_label_position("right")

    if label:
        ax.set_ylabel(label)

    if remove_spines:
        for spine in ax.spines.values():
            spine.set_visible(False)

    if separator and k > 1:
        if separator_kwargs is None:
            separator_kwargs = dict(colors="white", linewidth=0.5)
        ax.vlines(np.arange(1, k), vmin, vmax, **separator_kwargs)

    return ax
```

```python
# result = (
#     order_params
#     .sort_values("t", ascending=True)
#     .query("iha_tot > 0.45")
#     .groupby(["K", "seed", "n"], as_index=False)
#     .first()
#     .rename(columns={"t": "t_iha_threshold"})
#     [["K", "seed", "n", "iha_tot", "t_iha_threshold"]]
# )

threshold = 0.45
default_t = 1e4

def first_threshold_time(g):
    g = g.sort_values("t")
    mask = g["iha_tot"] > threshold
    
    if mask.any():
        row = g.loc[mask].iloc[0]
        return pd.Series({
            "iha_tot": row["iha_tot"],
            "t_iha_threshold": row["t"]
        })
    else:
        return pd.Series({
            "iha_tot": g["iha_tot"].iloc[-1],  # or np.nan if preferred
            "t_iha_threshold": default_t
        })

result = (
    order_params
    .groupby(["K", "seed", "n"], as_index=False)
    .apply(first_threshold_time)
    .reset_index()
)

t_bigen = pd.merge(k1, result, on = ['seed', 'n'])
t_bigen['K_diff'] = t_bigen['K'] - t_bigen['K1*']
t_bigen['K_diff_mean'] = t_bigen['K'] - t_bigen['K1*'].mean()
t_bigen['K_diff_median'] = t_bigen['K'] - t_bigen['K1*'].median()
```

```python
threshold = 0.45
default_t = 1e4

def first_threshold_time(g):
    g = g.sort_values("t")
    mask = g["iha_tot"] > threshold
    
    if mask.any():
        row = g.loc[mask].iloc[0]
        return pd.Series({
            "iha_tot": row["iha_tot"],
            "t_iha_threshold": row["t"]
        })
    else:
        return pd.Series({
            "iha_tot": g["iha_tot"].iloc[-1],  # or np.nan if preferred
            "t_iha_threshold": default_t
        })

result = (
    order_params
    .groupby(["K", "seed", "n"], as_index=False)
    .apply(first_threshold_time)
    .reset_index()
)
```

#### learning curves

```python
ax = sns.lineplot(model_perf[model_perf['step'] < 10**4], x = 'step', y = 'loss', hue = 'K', style = 'seed', palette = 'viridis', legend = False)
ax.set_xscale('log')
```

```python
def f(x, a: float, b: float, c: float, x0: float):
    return a+c*x0**b/(x**b + x0**b)

def x0(xx, yy):
    """ Estimate convergence time """
    half = yy.min() + (yy.max() - yy.min())/2
    idx = np.where(yy < half)[0]
    return xx[idx[0]].item()
```

```python
out = {
    't_ret': [],
    'converged': [],
    'N': [],
    param_name: [],
    'seed': [],
    'params': []
}

# for seed_idx in range(n_seeds):
for seed_idx in range(n_seeds):
    seed = exp_config.seedrange[seed_idx]
    for n_idx in range(n_n):
        n = exp_config.nrange[n_idx]
        for param_idx in range(n_params):
            param = exp_config.paramrange[param_idx]
            df = mask_df(model_perf, exp_config, seed_idx, n_idx, param_idx, 'K')
            if df.empty:
                print(f'pass param {param_idx}, n_idx {n_idx}')
                continue
            mask = (df['rolling_loss'][::-1].cummax()[::-1] < np.log(10))
            s_last = df.loc[mask, 'step'].min()
            df = df[df['step'] > s_last]
            df['s'] = df['step'].apply(lambda x: float(x))
            df['rolling'] = df['loss'].rolling(20, min_periods=1).mean()
            max_s = df['s'].max()
            xx = np.logspace(np.log10(s_last), np.log10(max_s), 100, base = 10)
            yy = np.interp(xx, df['s'], df['rolling'])
            # params, _ = curve_fit(f, xx.tolist(), yy.tolist(), maxfev = int(1e7),
            #                       bounds = ([-np.inf, 0.0, 0.0, 0.0], [np.inf, np.inf, np.inf, np.inf]),
            #                       p0 = [1, 1, 1, x0(xx, yy)]
            #                       )

            # optimal = mask_df(uni_ret, exp_config, seed_idx, n_idx, param_idx, param_name)['loss'].unique()
            params, _ = curve_fit(f, xx.tolist(), yy.tolist(), maxfev = int(1e7),
                        bounds = ([-np.inf, 0.0, 0.0, 0.0], [np.inf, np.inf, np.inf, np.inf]),
                        p0 = [1, 1, 1, x0(xx, yy)]
                        )
            out['t_ret'].append(params[-1])
            out['converged'].append(params[0])
            out['N'].append(n)
            out['seed'].append(seed)
            out[param_name].append(param)
            out['params'].append(params.tolist())

out = pd.DataFrame(out)
```

```python
import numpy as np
import pandas as pd

run_keys = ['K', 'seed', 'n']

model_df = model_perf.copy()
model_df = model_df.pivot_table(index=run_keys + ['t'], columns='dist', values='loss')

est_df = est_perf.copy()
est_df = est_df[est_df['name'].eq('2_gen')]
est_df = est_df.pivot_table(index=run_keys, columns='dist', values='loss')

merge_df = pd.merge(
    model_df.reset_index(),
    est_df.reset_index(),
    on=run_keys,
    suffixes=('_model', '_est')
).copy()

# y(t) = train_model(t) - train_est
merge_df['y'] = merge_df['train_model'] - merge_df['train_est']

# sort within runs
df = merge_df.sort_values(run_keys + ['t']).copy()

# previous point within run
df['t_prev'] = df.groupby(run_keys)['t'].shift(1)
df['y_prev'] = df.groupby(run_keys)['y'].shift(1)

# require positive times for log interpolation
valid_time = df['t_prev'].gt(0) & df['t'].gt(0)

# downward crossing: y_prev >= 0 and y < 0
# cross = valid_time & df['y_prev'].ge(0) & df['y'].lt(-1e-4)
cross = valid_time & df['y_prev'].ge(0) & df['y'].lt(0)

# log-time coordinates
df['log_t'] = np.log(df['t'])
df['log_t_prev'] = np.log(df['t_prev'])

# interpolate linearly in log t space:
# y = y_prev + (log_t_cross - log_t_prev) * (y - y_prev)/(log_t - log_t_prev)
# solve for log_t_cross, then exponentiate
df.loc[cross, 'log_t_cross'] = (
    df.loc[cross, 'log_t_prev']
    + (0.0 - df.loc[cross, 'y_prev'])
      * (df.loc[cross, 'log_t'] - df.loc[cross, 'log_t_prev'])
      / (df.loc[cross, 'y'] - df.loc[cross, 'y_prev'])
)

df.loc[cross, 't_cross'] = np.exp(df.loc[cross, 'log_t_cross'])

t_trans = (
    df.loc[cross]
      .groupby(run_keys, as_index=False)['t_cross']
      .min()
      .rename(columns={'t_cross': 't_trans'})
)
```

```python
fig, ax = plt.subplots(figsize=(1.75, 1.5))

# subtract t_bigen from t_transience
merged = pd.merge(t_trans, t_bigen, on=run_keys)
merged['t_persistent'] = merged['t_trans'] - merged['t_bigen']

N = 8 # number of seeds per run_key

# reasonable K range with interpretable dynamics
df_filtered = (
    merged[merged['K'].gt(2**7)]
    .sort_values('seed')
    .groupby(['K', 'n'], group_keys=False)
    .head(N)
)

df_fit = (
    df_filtered[df_filtered['K'].gt(2**9)]
    .groupby(['K', 'n'])['t_persistent']
    .median()
    .reset_index()
)

x = df_fit['K'].values
y = df_fit['t_persistent'].values

mask = (x > 0) & (y > 0)
x, y = x[mask], y[mask]

logx = np.log(x)
logy = np.log(y)

b, loga = np.polyfit(logx, logy, 1)
a = np.exp(loga)

# x_fit = np.linspace(x.min(), x.max(), 200)
x_fit = np.linspace(2**8, x.max(), 200)
y_fit = a * x_fit**b

print(b)

sns.violinplot(df_filtered, x='K', y='t_persistent',
                    native_scale=True,
                    color='C4',
                    log_scale=True,
                    ax=ax,
                    linewidth=0.75,
                    density_norm='width',
                    label='Observed',
)

# sns.boxplot(df_filtered, x='K', y='t_persistent',
#                     native_scale=True,
#                     color='C4',
#                     log_scale=True,
#                     ax=ax,
#                     linewidth=1,
#                     label='Observed',
#                     flierprops=dict(markersize=3),
# )

ax.plot(
    x_fit,
    y_fit, 
    color='black', 
    linestyle='--',
    linewidth=1,
    label=r'$\propto K^{\gamma}$',
    )

# only include first two legend entries
handles, labels = ax.get_legend_handles_labels()
ax.legend(handles[-2:], labels[-2:])

ax.set_ylabel(r'$\tau_{\mathrm{2-Mem}} - \tau_{\mathrm{2-Gen}}$')
ax.set_xlabel(r'$K$')

# ax.plot(2**13, 6e6, marker='x', color='red', label='')

# ax.axvline(5e3, color='black', linestyle=':', linewidth=1)

# plt.savefig(plot_dir / 'trans_dynamics.svg')
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit


def fit_divergence(
    x,
    y,
    ax,
    *,
    sigma=None,
    eps_frac=1e-12,
    ngrid=600,
    p0=None,
    label_data="data",
    label_fit="divergence fit",
    # --- divergence options ---
    div_form="shifted",   # "simple" : y = A * (x0 - x)^(-b)
                          # "shifted": y = y_inf + A * (x0 - x)^(-b)
    p0_div=None,
    # --- exponential options ---
    do_exp_fit=True,
    exp_form="shifted",   # "simple" : y = C * exp(k x)
                          # "shifted": y = y_inf + C * exp(k x)
    p0_exp=None,
    label_exp="exp fit",
    # --- NEW: plotting-domain options ---
    plot_extend=(0.0, 0.0),   # (left_frac, right_frac) of data span
    plot_xlim=None,           # (xmin_plot, xmax_plot) absolute; overrides plot_extend
    plot_clip_to_x0=True,     # avoid plotting past the fitted singularity x0
    x0_margin_frac=1e-6,      # margin relative to max(1, span) when clipping to x0
):
    """
    Divergence forms:
      - "simple" : y = A * (x0 - x)^(-b), with A>0, b>0, x0>max(x)
      - "shifted": y = y_inf + A * (x0 - x)^(-b), with A>0, b>0, x0>max(x), y_inf free

    Exponential forms:
      - "simple" : y = C * exp(k x), with C>0
      - "shifted": y = y_inf + C * exp(k x), with C>0, y_inf free
    """
    x = np.asarray(x, float).ravel()
    y = np.asarray(y, float).ravel()
    if x.shape != y.shape:
        raise ValueError("x and y must have the same shape")

    m = np.isfinite(x) & np.isfinite(y)

    if sigma is not None:
        sigma = np.asarray(sigma, float).ravel()
        if sigma.shape != y.shape:
            raise ValueError("sigma must have same shape as y")
        m &= np.isfinite(sigma) & (sigma > 0)
        sigma_fit = sigma[m]
    else:
        sigma_fit = None

    x_d = x[m]
    y_d = y[m]
    if x_d.size < 3:
        raise ValueError("Need at least 3 valid points")

    xmin, xmax = x_d.min(), x_d.max()
    xspan = xmax - xmin
    eps = eps_frac * max(xspan, 1.0)

    # ---- divergence model ----
    if div_form == "simple":
        def model_div(xv, A, x0, b):
            return A * (x0 - xv + eps) ** (-b)

        if p0_div is None:
            A0 = max(np.max(y_d), 1e-12)
            x0_0 = xmax + 0.2 * max(xspan, 1.0)
            b0 = 1.0
            p0_div = (A0, x0_0, b0)

        bounds_div = (
            (0.0, xmax + 1e-12, 0.0),
            (np.inf, np.inf, np.inf),
        )

    elif div_form == "shifted":
        def model_div(xv, y_inf, A, x0, b):
            return y_inf + A * (x0 - xv + eps) ** (-b)

        if p0_div is None:
            yinf0 = np.min(y_d)
            A0 = max(np.max(y_d) - yinf0, 1e-12)
            x0_0 = xmax + 0.2 * max(xspan, 1.0)
            b0 = 1.0
            p0_div = (yinf0, A0, x0_0, b0)

        bounds_div = (
            (-np.inf, 0.0, xmax + 1e-12, 0.0),
            ( np.inf, np.inf, np.inf, np.inf),
        )
    else:
        raise ValueError("div_form must be 'simple' or 'shifted'")

    popt_div, pcov_div = curve_fit(
        model_div,
        x_d,
        y_d,
        p0=p0_div if p0 is None else p0,
        bounds=bounds_div,
        sigma=sigma_fit,
        absolute_sigma=(sigma_fit is not None),
        maxfev=20000,
    )

    # ---- plotting data ----
    # ax.scatter(x_d, y_d, s=3, label=label_data)
    # sns.pointplot(x=x_d, y=y_d, color='C0', ax=ax, markersize=1, estimator='median', errorbar='ci', label=label_data, native_scale=True, linewidth=1)

    # ---- NEW: choose plotting domain ----
    if plot_xlim is not None:
        xmin_plot, xmax_plot = map(float, plot_xlim)
        if not (np.isfinite(xmin_plot) and np.isfinite(xmax_plot) and xmin_plot < xmax_plot):
            raise ValueError("plot_xlim must be (xmin_plot, xmax_plot) with xmin_plot < xmax_plot")
    else:
        left_frac, right_frac = plot_extend
        if not (np.isfinite(left_frac) and np.isfinite(right_frac)):
            raise ValueError("plot_extend must be (left_frac, right_frac) finite numbers")
        span = max(xspan, 1.0)
        xmin_plot = xmin - left_frac * span
        xmax_plot = xmax + right_frac * span

    # Clip curve to remain left of the fitted x0 (avoid crossing the singularity)
    if plot_clip_to_x0:
        if div_form == "simple":
            _, x0_hat, _ = popt_div
        else:
            _, _, x0_hat, _ = popt_div
        margin = x0_margin_frac * max(xspan, 1.0)
        xmax_plot = min(xmax_plot, x0_hat - margin)

    xx = np.linspace(xmin_plot, xmax_plot, ngrid)
    yy_div = model_div(xx, *popt_div)

    if div_form == "simple":
        A_hat, x0_hat, b_hat = popt_div
        div_label = (
            rf"{label_fit}: $A(x_0-x)^{{-b}}$, "
            rf"$x_0={x0_hat:.6g}$, $b={b_hat:.3g}$"
        )
    else:
        yinf_hat, A_hat, x0_hat, b_hat = popt_div
        div_label = (
            rf"{label_fit}: $y_\infty + A(x_0-x)^{{-b}}$, "
            rf"$y_\infty={yinf_hat:.6g}$, $x_0={x0_hat:.6g}$, $b={b_hat:.3g}$"
        )

    ax.plot(xx, yy_div, linewidth=1, label=div_label, color='black', linestyle='--')

    # ---- optional exponential fit ----
    popt_exp, pcov_exp = None, None
    if do_exp_fit:
        if exp_form == "simple":
            def model_exp(xv, C, k):
                return C * np.exp(k * xv)

            if p0_exp is None:
                y_pos = y_d[y_d > 0]
                C0 = np.median(y_pos) if y_pos.size >= 2 else max(np.median(y_d), 1e-12)
                k0 = 0.0
                p0_exp = (C0, k0)

            bounds_exp = ((0.0, -np.inf), (np.inf, np.inf))

        elif exp_form == "shifted":
            def model_exp(xv, y_inf, C, k):
                return y_inf + C * np.exp(k * xv)

            if p0_exp is None:
                y_inf0 = np.min(y_d)
                C0 = max(np.max(y_d) - y_inf0, 1e-12)
                k0 = 0.0
                p0_exp = (y_inf0, C0, k0)

            bounds_exp = ((-np.inf, 0.0, -np.inf), (np.inf, np.inf, np.inf))
        else:
            raise ValueError("exp_form must be 'simple' or 'shifted'")

        popt_exp, pcov_exp = curve_fit(
            model_exp,
            x_d,
            y_d,
            p0=p0_exp,
            bounds=bounds_exp,
            sigma=sigma_fit,
            absolute_sigma=(sigma_fit is not None),
            maxfev=20000,
        )

        yy_exp = model_exp(xx, *popt_exp)

        if exp_form == "simple":
            C_hat, k_hat = popt_exp
            exp_label = rf"{label_exp}: $C e^{{kx}}$, $k={k_hat:.3g}$"
        else:
            yinf_hat, C_hat, k_hat = popt_exp
            exp_label = (
                rf"{label_exp}: $y_\infty + C e^{{kx}}$, "
                rf"$y_\infty={yinf_hat:.6g}$, $k={k_hat:.3g}$"
            )

        ax.plot(xx, yy_exp, linewidth=1, linestyle="--", label=exp_label)

    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.legend()

    return (popt_div, pcov_div), (popt_exp, pcov_exp)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit


def _power_law_xc_no_shift_left(x, A, gamma, x_c):
    t = x_c - x
    return A * np.power(t, -gamma)


def _power_law_xc_with_shift_left(x, A, gamma, B, x_c):
    t = x_c - x
    return A * np.power(t, -gamma) + B


def _power_law_xc_no_shift_right(x, A, gamma, x_c):
    t = x - x_c
    return A * np.power(t, -gamma)


def _power_law_xc_with_shift_right(x, A, gamma, B, x_c):
    t = x - x_c
    return A * np.power(t, -gamma) + B


def _safe_stderr_from_pcov(pcov, n_params):
    if pcov is None:
        return np.full(n_params, np.nan)
    pcov = np.asarray(pcov)
    if pcov.shape != (n_params, n_params):
        return np.full(n_params, np.nan)
    diag = np.diag(pcov)
    diag = np.where(diag >= 0, diag, np.nan)
    return np.sqrt(diag)


def _weighted_rss(y_true, y_pred, sigma=None):
    resid = y_true - y_pred
    if sigma is None:
        return np.sum(resid**2)
    return np.sum((resid / sigma) ** 2)


def _reduced_chi2(y_true, y_pred, n_params, sigma=None):
    dof = len(y_true) - n_params
    if dof <= 0:
        return np.nan
    return _weighted_rss(y_true, y_pred, sigma=sigma) / dof


def _corr_from_pcov(pcov):
    if pcov is None:
        return None
    d = np.sqrt(np.diag(pcov))
    with np.errstate(invalid="ignore", divide="ignore"):
        corr = pcov / np.outer(d, d)
    return corr


def _parameter_names(side, model):
    if side == "left" and model == "no_shift":
        return ["A", "gamma", "x_c"]
    if side == "left" and model == "with_shift":
        return ["A", "gamma", "B", "x_c"]
    if side == "right" and model == "no_shift":
        return ["A", "gamma", "x_c"]
    if side == "right" and model == "with_shift":
        return ["A", "gamma", "B", "x_c"]
    raise ValueError("Invalid side/model combination.")


def _get_model_function(side, model):
    if side == "left" and model == "no_shift":
        return _power_law_xc_no_shift_left, 3
    if side == "left" and model == "with_shift":
        return _power_law_xc_with_shift_left, 4
    if side == "right" and model == "no_shift":
        return _power_law_xc_no_shift_right, 3
    if side == "right" and model == "with_shift":
        return _power_law_xc_with_shift_right, 4
    raise ValueError("side must be 'left' or 'right'; model must be 'no_shift' or 'with_shift'.")


def _default_initial_guess_and_bounds(x, y, side, model, x_c_guess=None):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    if x_c_guess is None:
        if side == "left":
            x_c_guess = np.max(x) + 0.05 * (np.max(x) - np.min(x) + 1e-12)
        elif side == "right":
            x_c_guess = np.min(x) - 0.05 * (np.max(x) - np.min(x) + 1e-12)
        else:
            raise ValueError("side must be 'left' or 'right'.")

    if side == "left":
        t0 = x_c_guess - x
    else:
        t0 = x - x_c_guess

    good = t0 > 0
    if not np.any(good):
        raise ValueError("Initial x_c_guess places all points outside the valid fitting region.")

    xg = x[good]
    yg = y[good]
    tg = t0[good]

    gamma0 = 1.0
    B0 = np.min(yg)
    A0 = (np.max(yg) - np.min(yg) + 1e-12) * np.power(np.median(tg), gamma0)

    x_span = np.max(x) - np.min(x)
    pad = max(0.1 * x_span, 1e-8)

    if side == "left":
        x_c_lo = np.max(x) + 1e-12
        x_c_hi = np.max(x) + 5.0 * x_span + pad
    else:
        x_c_lo = np.min(x) - 5.0 * x_span - pad
        x_c_hi = np.min(x) - 1e-12

    if model == "no_shift":
        p0 = [A0, gamma0, x_c_guess]
        bounds = ([-np.inf, -np.inf, x_c_lo], [np.inf, np.inf, x_c_hi])
    else:
        p0 = [A0, gamma0, B0, x_c_guess]
        bounds = ([-np.inf, -np.inf, -np.inf, x_c_lo], [np.inf, np.inf, np.inf, x_c_hi])

    return p0, bounds


def _fit_single_window_free_xc(
    x,
    y,
    side="left",
    model="no_shift",
    sigma=None,
    p0=None,
    bounds=None,
    x_c_guess=None,
    maxfev=50000,
):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    if sigma is not None:
        sigma = np.asarray(sigma, dtype=float)

    f, n_params = _get_model_function(side, model)

    if p0 is None or bounds is None:
        p0_default, bounds_default = _default_initial_guess_and_bounds(
            x=x, y=y, side=side, model=model, x_c_guess=x_c_guess
        )
        if p0 is None:
            p0 = p0_default
        if bounds is None:
            bounds = bounds_default

    popt, pcov = curve_fit(
        f,
        x,
        y,
        p0=p0,
        sigma=sigma,
        absolute_sigma=(sigma is not None),
        bounds=bounds,
        maxfev=maxfev,
    )
    perr = _safe_stderr_from_pcov(pcov, n_params)
    y_fit = f(x, *popt)

    fit = {
        "params": popt,
        "param_stderr": perr,
        "pcov": pcov,
        "corr": _corr_from_pcov(pcov),
        "y_fit": y_fit,
        "rss": _weighted_rss(y, y_fit, sigma=sigma),
        "reduced_chi2": _reduced_chi2(y, y_fit, n_params=n_params, sigma=sigma),
        "n_points": len(x),
        "side": side,
        "model": model,
        "parameter_names": _parameter_names(side, model),
    }

    if model == "no_shift":
        fit["A"], fit["gamma"], fit["x_c"] = popt
        fit["A_stderr"], fit["gamma_stderr"], fit["x_c_stderr"] = perr
        fit["B"] = np.nan
        fit["B_stderr"] = np.nan
        fit["corr_gamma_xc"] = fit["corr"][1, 2] if fit["corr"] is not None else np.nan
        fit["corr_B_gamma"] = np.nan
        fit["corr_B_xc"] = np.nan
    else:
        fit["A"], fit["gamma"], fit["B"], fit["x_c"] = popt
        fit["A_stderr"], fit["gamma_stderr"], fit["B_stderr"], fit["x_c_stderr"] = perr
        fit["corr_gamma_xc"] = fit["corr"][1, 3] if fit["corr"] is not None else np.nan
        fit["corr_B_gamma"] = fit["corr"][2, 1] if fit["corr"] is not None else np.nan
        fit["corr_B_xc"] = fit["corr"][2, 3] if fit["corr"] is not None else np.nan

    return fit


def exponent_stability_scan_free_xc(
    x,
    y,
    sigma_y=None,
    side="left",
    model="no_shift",
    min_points=6,
    max_points=None,
    sort_closest_to_edge=True,
    x_c_guess=None,
    p0=None,
    bounds=None,
    maxfev=50000,
    reuse_previous_fit_as_init=True,
):
    """
    Nested-window exponent stability scan with x_c fitted as a parameter.

    Parameters
    ----------
    x, y : 1D arrays
        Data.
    sigma_y : 1D array or None
        Optional y uncertainties.
    side : {'left', 'right'}
        Which side of the singularity the data lie on.
        'left' means x < x_c.
        'right' means x > x_c.
    model : {'no_shift', 'with_shift'}
        Fit A * |x - x_c|^{-gamma}  or  A * |x - x_c|^{-gamma} + B.
    min_points, max_points : int
        Range of nested window sizes.
    sort_closest_to_edge : bool
        If True, sort by distance to the relevant data edge:
        - left: closest to max(x)
        - right: closest to min(x)
        This is the sensible default when x_c is not yet known.
    x_c_guess : float or None
        Initial guess for x_c.
    p0, bounds : optional
        Passed to curve_fit. Usually leave as None.
    reuse_previous_fit_as_init : bool
        Reuse previous window fit as the initial guess for the next window.

    Returns
    -------
    results : dict
        Arrays of fitted parameters and diagnostics for each window.
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    if x.ndim != 1 or y.ndim != 1 or len(x) != len(y):
        raise ValueError("x and y must be 1D arrays of equal length.")

    if side not in ("left", "right"):
        raise ValueError("This free-x_c version only supports side='left' or side='right'.")

    if sigma_y is not None:
        sigma_y = np.asarray(sigma_y, dtype=float)
        if sigma_y.shape != y.shape:
            raise ValueError("sigma_y must have the same shape as y.")
        if np.any(sigma_y <= 0):
            raise ValueError("All sigma_y values must be positive.")

    finite_mask = np.isfinite(x) & np.isfinite(y)
    if sigma_y is not None:
        finite_mask &= np.isfinite(sigma_y)

    x = x[finite_mask]
    y = y[finite_mask]
    if sigma_y is not None:
        sigma_y = sigma_y[finite_mask]

    if len(x) < min_points:
        raise ValueError("Not enough finite points.")

    if sort_closest_to_edge:
        if side == "left":
            order = np.argsort(np.max(x) - x)
        else:
            order = np.argsort(x - np.min(x))
    else:
        order = np.argsort(x)

    x = x[order]
    y = y[order]
    if sigma_y is not None:
        sigma_y = sigma_y[order]

    if max_points is None:
        max_points = len(x)
    max_points = min(max_points, len(x))

    if max_points < min_points:
        raise ValueError("max_points must be >= min_points.")

    out = {
        "n_points": [],
        "x_min": [],
        "x_max": [],
        "A": [],
        "A_stderr": [],
        "gamma": [],
        "gamma_stderr": [],
        "B": [],
        "B_stderr": [],
        "x_c": [],
        "x_c_stderr": [],
        "rss": [],
        "reduced_chi2": [],
        "corr_gamma_xc": [],
        "corr_B_gamma": [],
        "corr_B_xc": [],
        "success": [],
        "fit_objects": [],
    }

    prev_p0 = p0
    for m in range(min_points, max_points + 1):
        x_win = x[:m]
        y_win = y[:m]
        sigma_win = sigma_y[:m] if sigma_y is not None else None

        try:
            fit = _fit_single_window_free_xc(
                x=x_win,
                y=y_win,
                side=side,
                model=model,
                sigma=sigma_win,
                p0=prev_p0 if reuse_previous_fit_as_init else p0,
                bounds=bounds,
                x_c_guess=x_c_guess,
                maxfev=maxfev,
            )
            success = True
            if reuse_previous_fit_as_init:
                prev_p0 = fit["params"]
        except Exception:
            fit = None
            success = False

        out["n_points"].append(m)
        out["x_min"].append(np.min(x_win))
        out["x_max"].append(np.max(x_win))
        out["success"].append(success)
        out["fit_objects"].append(fit)

        if success:
            out["A"].append(fit["A"])
            out["A_stderr"].append(fit["A_stderr"])
            out["gamma"].append(fit["gamma"])
            out["gamma_stderr"].append(fit["gamma_stderr"])
            out["B"].append(fit["B"])
            out["B_stderr"].append(fit["B_stderr"])
            out["x_c"].append(fit["x_c"])
            out["x_c_stderr"].append(fit["x_c_stderr"])
            out["rss"].append(fit["rss"])
            out["reduced_chi2"].append(fit["reduced_chi2"])
            out["corr_gamma_xc"].append(fit["corr_gamma_xc"])
            out["corr_B_gamma"].append(fit["corr_B_gamma"])
            out["corr_B_xc"].append(fit["corr_B_xc"])
        else:
            out["A"].append(np.nan)
            out["A_stderr"].append(np.nan)
            out["gamma"].append(np.nan)
            out["gamma_stderr"].append(np.nan)
            out["B"].append(np.nan)
            out["B_stderr"].append(np.nan)
            out["x_c"].append(np.nan)
            out["x_c_stderr"].append(np.nan)
            out["rss"].append(np.nan)
            out["reduced_chi2"].append(np.nan)
            out["corr_gamma_xc"].append(np.nan)
            out["corr_B_gamma"].append(np.nan)
            out["corr_B_xc"].append(np.nan)

    for key in out:
        if key != "fit_objects":
            out[key] = np.asarray(out[key])

    out["side"] = side
    out["model"] = model
    return out


def compare_shift_models_free_xc(
    x,
    y,
    sigma_y=None,
    side="left",
    min_points=6,
    max_points=None,
    x_c_guess=None,
    p0_no_shift=None,
    p0_with_shift=None,
    bounds_no_shift=None,
    bounds_with_shift=None,
    maxfev=50000,
):
    """
    Run nested-window scans for both models with x_c free.
    """
    res_no = exponent_stability_scan_free_xc(
        x=x,
        y=y,
        sigma_y=sigma_y,
        side=side,
        model="no_shift",
        min_points=min_points,
        max_points=max_points,
        x_c_guess=x_c_guess,
        p0=p0_no_shift,
        bounds=bounds_no_shift,
        maxfev=maxfev,
    )
    res_shift = exponent_stability_scan_free_xc(
        x=x,
        y=y,
        sigma_y=sigma_y,
        side=side,
        model="with_shift",
        min_points=min_points,
        max_points=max_points,
        x_c_guess=x_c_guess,
        p0=p0_with_shift,
        bounds=bounds_with_shift,
        maxfev=maxfev,
    )
    return res_no, res_shift


def plot_stability_results_free_xc(res_no_shift, res_with_shift=None):
    """
    Plot gamma, x_c, B, and reduced chi^2 versus window size.
    """
    nrows = 4 if res_with_shift is None else 5
    fig, axes = plt.subplots(nrows, 1, figsize=(7, 3.0 * nrows), sharex=True)

    # gamma
    axes[0].errorbar(
        res_no_shift["n_points"],
        res_no_shift["gamma"],
        yerr=res_no_shift["gamma_stderr"],
        fmt="o-",
        label="no shift",
    )
    if res_with_shift is not None:
        axes[0].errorbar(
            res_with_shift["n_points"],
            res_with_shift["gamma"],
            yerr=res_with_shift["gamma_stderr"],
            fmt="o-",
            label="with shift",
        )
    axes[0].set_ylabel(r"$\gamma$")
    axes[0].grid(True, alpha=0.3)
    axes[0].legend()

    # x_c
    axes[1].errorbar(
        res_no_shift["n_points"],
        res_no_shift["x_c"],
        yerr=res_no_shift["x_c_stderr"],
        fmt="o-",
        label="no shift",
    )
    if res_with_shift is not None:
        axes[1].errorbar(
            res_with_shift["n_points"],
            res_with_shift["x_c"],
            yerr=res_with_shift["x_c_stderr"],
            fmt="o-",
            label="with shift",
        )
    axes[1].set_ylabel(r"$x_c$")
    axes[1].grid(True, alpha=0.3)
    axes[1].legend()

    # gamma-xc correlation
    axes[2].plot(
        res_no_shift["n_points"],
        res_no_shift["corr_gamma_xc"],
        "o-",
        label="no shift",
    )
    if res_with_shift is not None:
        axes[2].plot(
            res_with_shift["n_points"],
            res_with_shift["corr_gamma_xc"],
            "o-",
            label="with shift",
        )
    axes[2].set_ylabel(r"corr$(\gamma,x_c)$")
    axes[2].grid(True, alpha=0.3)
    axes[2].legend()

    row = 3
    if res_with_shift is not None:
        axes[row].errorbar(
            res_with_shift["n_points"],
            res_with_shift["B"],
            yerr=res_with_shift["B_stderr"],
            fmt="o-",
            label="B",
        )
        axes[row].set_ylabel("shift B")
        axes[row].grid(True, alpha=0.3)
        axes[row].legend()
        row += 1

    axes[row].plot(
        res_no_shift["n_points"],
        res_no_shift["reduced_chi2"],
        "o-",
        label="no shift",
    )
    if res_with_shift is not None:
        axes[row].plot(
            res_with_shift["n_points"],
            res_with_shift["reduced_chi2"],
            "o-",
            label="with shift",
        )
    axes[row].set_xlabel("number of points in nested window")
    axes[row].set_ylabel(r"reduced $\chi^2$")
    axes[row].grid(True, alpha=0.3)
    axes[row].legend()

    fig.tight_layout()
    return fig, axes


def plot_representative_fits_free_xc(
    x,
    y,
    results,
    sigma_y=None,
    n_windows=(0, -1),
    n_grid=400,
    loglog=False,
):
    """
    Plot data and selected fitted curves for representative windows.

    Since x_c varies by window, plotting against x is cleaner than plotting
    against t. If loglog=True, the x-axis is left linear; only y is logged.
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    if sigma_y is not None:
        sigma_y = np.asarray(sigma_y, dtype=float)

    side = results["side"]
    if side == "left":
        order = np.argsort(np.max(x) - x)
    else:
        order = np.argsort(x - np.min(x))

    x_sorted = x[order]
    y_sorted = y[order]
    sigma_sorted = sigma_y[order] if sigma_y is not None else None

    fig, ax = plt.subplots(figsize=(7, 5))
    if sigma_sorted is None:
        ax.plot(x_sorted, y_sorted, "o", label="data")
    else:
        ax.errorbar(x_sorted, y_sorted, yerr=sigma_sorted, fmt="o", label="data")

    model = results["model"]
    for idx in n_windows:
        fit = results["fit_objects"][idx]
        if fit is None:
            continue

        m = results["n_points"][idx]
        x_win = x_sorted[:m]

        if side == "left":
            x_lo = np.min(x_win)
            x_hi = min(np.max(x_win), fit["x_c"] - 1e-12)
        else:
            x_lo = max(np.min(x_win), fit["x_c"] + 1e-12)
            x_hi = np.max(x_win)

        if not np.isfinite(x_lo) or not np.isfinite(x_hi) or x_hi <= x_lo:
            continue

        x_grid = np.linspace(x_lo, x_hi, n_grid)

        if side == "left" and model == "no_shift":
            y_grid = _power_law_xc_no_shift_left(x_grid, fit["A"], fit["gamma"], fit["x_c"])
            label = f"m={m}, gamma={fit['gamma']:.4g}, x_c={fit['x_c']:.6g}"
        elif side == "left" and model == "with_shift":
            y_grid = _power_law_xc_with_shift_left(x_grid, fit["A"], fit["gamma"], fit["B"], fit["x_c"])
            label = f"m={m}, gamma={fit['gamma']:.4g}, B={fit['B']:.4g}, x_c={fit['x_c']:.6g}"
        elif side == "right" and model == "no_shift":
            y_grid = _power_law_xc_no_shift_right(x_grid, fit["A"], fit["gamma"], fit["x_c"])
            label = f"m={m}, gamma={fit['gamma']:.4g}, x_c={fit['x_c']:.6g}"
        else:
            y_grid = _power_law_xc_with_shift_right(x_grid, fit["A"], fit["gamma"], fit["B"], fit["x_c"])
            label = f"m={m}, gamma={fit['gamma']:.4g}, B={fit['B']:.4g}, x_c={fit['x_c']:.6g}"

        ax.plot(x_grid, y_grid, "-", label=label)

    ax.set_xlabel("x")
    ax.set_ylabel("y")
    if loglog:
        ax.set_yscale("log")
    ax.grid(True, alpha=0.3)
    ax.legend()
    fig.tight_layout()
    return fig, ax


def summarize_last_window(results):
    """
    Compact summary of the largest-window fit.
    """
    i = -1
    return {
        "model": results["model"],
        "side": results["side"],
        "n_points": int(results["n_points"][i]),
        "gamma": results["gamma"][i],
        "gamma_stderr": results["gamma_stderr"][i],
        "x_c": results["x_c"][i],
        "x_c_stderr": results["x_c_stderr"][i],
        "B": results["B"][i],
        "B_stderr": results["B_stderr"][i],
        "reduced_chi2": results["reduced_chi2"][i],
        "corr_gamma_xc": results["corr_gamma_xc"][i],
    }


# --------------------------
# Minimal usage on your data
# --------------------------
# Example:
#
# res_no, res_shift = compare_shift_models_free_xc(
#     x=x,
#     y=y,
#     sigma_y=None,          # or your y error bars
#     side="left",           # use "left" if you expect x < x_c
#     min_points=8,
#     max_points=25,
#     x_c_guess=xc_guess,    # important; provide a sensible guess
# )
#
# plot_stability_results_free_xc(res_no, res_shift)
# plt.show()
#
# print(summarize_last_window(res_no))
# print(summarize_last_window(res_shift))
#
# plot_representative_fits_free_xc(x, y, res_no, sigma_y=None, n_windows=(0, -1))
# plot_representative_fits_free_xc(x, y, res_shift, sigma_y=None, n_windows=(0, -1))
# plt.show()
```

```python
import numpy as np
import pandas as pd
from scipy.optimize import curve_fit


def hill_sat(t, A, tau, n):
    """
    Hill-type saturation curve with known baseline y(0)=0 and asymptote A <= 1.

    Parameters
    ----------
    t : array-like
        Nonnegative times.
    A : float
        Saturation value, constrained to [0, 1].
    tau : float
        Half-saturation timescale (> 0).
    n : float
        Hill coefficient (> 0).

    Returns
    -------
    ndarray
        Model values.
    """
    t = np.asarray(t, dtype=float)
    t_pow = np.power(t, n)
    tau_pow = tau ** n
    return A * t_pow / (t_pow + tau_pow)


def fit_group_weighted(
    df,
    tail_weight="logt",
    min_points=3,
    min_positive_t=2,
    maxfev=20000,
):
    """
    Fit a Hill saturation curve with:
      - known starting value y(0)=0
      - constrained saturation 0 <= A <= 1

    Parameters
    ----------
    df : pandas.DataFrame
        Must contain columns 't' and 'fraction'.
    tail_weight : {'logt', 't', None}, default 'logt'
        Weighting scheme to emphasize larger t.
    min_points : int, default 3
        Minimum number of finite points required.
    min_positive_t : int, default 2
        Minimum number of strictly positive t values required.
    maxfev : int, default 20000
        Maximum function evaluations for curve_fit.

    Returns
    -------
    pandas.Series
        Keys: A_fit, tau_fit, n_fit
    """
    g = df.sort_values("t").copy()
    t = g["t"].to_numpy(dtype=float)
    y = g["fraction"].to_numpy(dtype=float)

    mask = np.isfinite(t) & np.isfinite(y)
    t = t[mask]
    y = y[mask]

    if len(t) < min_points or np.sum(t > 0) < min_positive_t:
        return pd.Series({"A_fit": np.nan, "tau_fit": np.nan, "n_fit": np.nan})

    # The model is intended for nonnegative t and fractions in [0, 1].
    valid = (t >= 0) & (y >= 0) & (y <= 1)
    t = t[valid]
    y = y[valid]

    if len(t) < min_points or np.sum(t > 0) < min_positive_t:
        return pd.Series({"A_fit": np.nan, "tau_fit": np.nan, "n_fit": np.nan})

    t_pos = t[t > 0]

    # Initialization:
    # A0 should respect the known upper bound.
    A0 = float(np.clip(np.nanmax(y), 1e-6, 0.999))
    tau0 = float(np.median(t_pos))
    n0 = 2.0

    # Larger weight => smaller sigma in curve_fit
    if tail_weight == "logt":
        w = np.ones_like(t, dtype=float)
        pos = t > 0
        if np.any(pos):
            lt = np.log(t[pos])
            span = lt.max() - lt.min()
            if span > 0:
                lt_scaled = (lt - lt.min()) / span
            else:
                lt_scaled = np.zeros_like(lt)
            w[pos] = 1.0 + 4.0 * lt_scaled
    elif tail_weight == "t":
        w = np.ones_like(t, dtype=float)
        tmax = np.max(t)
        if tmax > 0:
            w = 1.0 + 4.0 * (t / tmax)
    else:
        w = np.ones_like(t, dtype=float)

    sigma = 1.0 / w

    try:
        popt, _ = curve_fit(
            hill_sat,
            t,
            y,
            p0=[A0, tau0, n0],
            bounds=([0.0, 1e-12, 1e-12], [1.0, np.inf, np.inf]),
            sigma=sigma,
            absolute_sigma=False,
            maxfev=maxfev,
        )
        A_fit, tau_fit, n_fit = popt
    except (RuntimeError, ValueError):
        A_fit, tau_fit, n_fit = np.nan, np.nan, np.nan

    return pd.Series({
        "A_fit": A_fit,
        "tau_fit": tau_fit,
        "n_fit": n_fit,
    })

fits = (
    neg_counts
    .groupby(["K", "seed"], group_keys=False)
    .apply(fit_group_weighted, tail_weight=None)
    .reset_index()
)

print(fits)

fig, ax = plt.subplots(figsize=(1.75, 1.5))
sns.pointplot(
    fits,
    x='K', 
    y='A_fit', 
    ax=ax, 
    estimator='mean', 
    errorbar='sd', 
    markersize=3, 
    capsize=0.5,
    linewidth=1, 
    log_scale=(True, False), 
    native_scale=True
    )

ax.set_xlabel(r'$K$')
ax.set_ylabel(r'Capacity (frac. $\mathcal{D}_T$)')

plt.savefig(plot_dir / 'mem_capacity.svg')
```

```python
import numpy as np
import pandas as pd


def variance_explained_by_task(df, y_col='gap', task_col='task', K_col='K'):
    rows = []

    for K, g in df.groupby(K_col):
        y = g[y_col].to_numpy()
        overall_mean = y.mean()

        ss_total = np.sum((y - overall_mean) ** 2)

        group_stats = g.groupby(task_col)[y_col].agg(['mean', 'size'])
        ss_between = np.sum(group_stats['size'] * (group_stats['mean'] - overall_mean) ** 2)

        r2 = ss_between / ss_total if ss_total > 0 else np.nan

        rows.append({
            'K': K,
            'n_obs': len(g),
            'n_tasks': g[task_col].nunique(),
            'ss_total': ss_total,
            'ss_between': ss_between,
            'variance_explained': r2,
        })

    return pd.DataFrame(rows).sort_values('K').reset_index(drop=True)
```

```python
import numpy as np
import pandas as pd
from scipy.optimize import curve_fit

# def hill_amp(t, A, tau, n):
#     t = np.asarray(t, dtype=float)
#     return A * (t**n) / (t**n + tau**n)

# def fit_group_weighted(df, tail_weight="logt"):
#     g = df.sort_values("t").copy()
#     t = g["t"].to_numpy(dtype=float)
#     y = g["fraction"].to_numpy(dtype=float)

#     mask = np.isfinite(t) & np.isfinite(y)
#     t = t[mask]
#     y = y[mask]

#     if len(t) < 3 or np.sum(t > 0) < 2:
#         return pd.Series({"A_fit": np.nan, "tau_fit": np.nan, "n_fit": np.nan})

#     t_pos = t[t > 0]

#     A0 = max(np.nanmax(y), 1e-6)
#     tau0 = np.median(t_pos)
#     n0 = 2.0

#     # Larger sigma = lower weight in curve_fit
#     if tail_weight == "logt":
#         # give more weight to large t by making sigma smaller there
#         w = np.ones_like(t)
#         pos = t > 0
#         lt = np.log(t[pos])
#         lt = (lt - lt.min()) / (lt.max() - lt.min() + 1e-12)
#         w[pos] = 1.0 + 4.0 * lt
#     elif tail_weight == "t":
#         w = 1.0 + 4.0 * (t / (t.max() + 1e-12))
#     else:
#         w = np.ones_like(t)

#     sigma = 1.0 / w

#     try:
#         popt, pcov = curve_fit(
#             hill_amp,
#             t,
#             y,
#             p0=[A0, tau0, n0],
#             bounds=([0.0, 1e-12, 1e-12], [np.inf, np.inf, np.inf]),
#             sigma=sigma,
#             absolute_sigma=False,
#             maxfev=20000,
#         )
#         A_fit, tau_fit, n_fit = popt
#     except RuntimeError:
#         A_fit, tau_fit, n_fit = np.nan, np.nan, np.nan

#     return pd.Series({
#         "A_fit": A_fit,
#         "tau_fit": tau_fit,
#         "n_fit": n_fit,
#     })

import numpy as np
import pandas as pd
from scipy.optimize import curve_fit


def hill_sat(t, A, tau, n):
    """
    Hill-type saturation curve with known baseline y(0)=0 and asymptote A <= 1.

    Parameters
    ----------
    t : array-like
        Nonnegative times.
    A : float
        Saturation value, constrained to [0, 1].
    tau : float
        Half-saturation timescale (> 0).
    n : float
        Hill coefficient (> 0).

    Returns
    -------
    ndarray
        Model values.
    """
    t = np.asarray(t, dtype=float)
    t_pow = np.power(t, n)
    tau_pow = tau ** n
    return A * t_pow / (t_pow + tau_pow)


def fit_group_weighted(
    df,
    tail_weight="logt",
    min_points=3,
    min_positive_t=2,
    maxfev=20000,
):
    """
    Fit a Hill saturation curve with:
      - known starting value y(0)=0
      - constrained saturation 0 <= A <= 1

    Parameters
    ----------
    df : pandas.DataFrame
        Must contain columns 't' and 'fraction'.
    tail_weight : {'logt', 't', None}, default 'logt'
        Weighting scheme to emphasize larger t.
    min_points : int, default 3
        Minimum number of finite points required.
    min_positive_t : int, default 2
        Minimum number of strictly positive t values required.
    maxfev : int, default 20000
        Maximum function evaluations for curve_fit.

    Returns
    -------
    pandas.Series
        Keys: A_fit, tau_fit, n_fit
    """
    g = df.sort_values("t").copy()
    t = g["t"].to_numpy(dtype=float)
    y = g["fraction"].to_numpy(dtype=float)

    mask = np.isfinite(t) & np.isfinite(y)
    t = t[mask]
    y = y[mask]

    if len(t) < min_points or np.sum(t > 0) < min_positive_t:
        return pd.Series({"A_fit": np.nan, "tau_fit": np.nan, "n_fit": np.nan})

    # The model is intended for nonnegative t and fractions in [0, 1].
    valid = (t >= 0) & (y >= 0) & (y <= 1)
    t = t[valid]
    y = y[valid]

    if len(t) < min_points or np.sum(t > 0) < min_positive_t:
        return pd.Series({"A_fit": np.nan, "tau_fit": np.nan, "n_fit": np.nan})

    t_pos = t[t > 0]

    # Initialization:
    # A0 should respect the known upper bound.
    A0 = float(np.clip(np.nanmax(y), 1e-6, 0.999))
    tau0 = float(np.median(t_pos))
    n0 = 2.0

    # Larger weight => smaller sigma in curve_fit
    if tail_weight == "logt":
        w = np.ones_like(t, dtype=float)
        pos = t > 0
        if np.any(pos):
            lt = np.log(t[pos])
            span = lt.max() - lt.min()
            if span > 0:
                lt_scaled = (lt - lt.min()) / span
            else:
                lt_scaled = np.zeros_like(lt)
            w[pos] = 1.0 + 4.0 * lt_scaled
    elif tail_weight == "t":
        w = np.ones_like(t, dtype=float)
        tmax = np.max(t)
        if tmax > 0:
            w = 1.0 + 4.0 * (t / tmax)
    else:
        w = np.ones_like(t, dtype=float)

    sigma = 1.0 / w

    try:
        popt, _ = curve_fit(
            hill_sat,
            t,
            y,
            p0=[A0, tau0, n0],
            bounds=([0.0, 1e-12, 1e-12], [1.0, np.inf, np.inf]),
            sigma=sigma,
            absolute_sigma=False,
            maxfev=maxfev,
        )
        A_fit, tau_fit, n_fit = popt
    except (RuntimeError, ValueError):
        A_fit, tau_fit, n_fit = np.nan, np.nan, np.nan

    return pd.Series({
        "A_fit": A_fit,
        "tau_fit": tau_fit,
        "n_fit": n_fit,
    })

fits = (
    neg_counts
    .groupby(["K", "seed"], group_keys=False)
    .apply(fit_group_weighted, tail_weight=None)
    .reset_index()
)

print(fits)

fig, ax = plt.subplots(figsize=(1.75, 1.5))
sns.pointplot(fits, x='K', y='A_fit', ax=ax, estimator='mean', errorbar='sd', markersize=3, linewidth=1, log_scale=(False, False), native_scale=True)

ax.set_xlabel(r'$K$')
ax.set_ylabel('Fit Asymptote')
```

```python
import numpy as np
import pandas as pd

run_keys = ['K', 'seed', 'n']

model_df = model_perf.copy()
model_df = model_df.pivot_table(index=run_keys + ['t'], columns='dist', values='loss')

est_df = est_perf.copy()
est_df = est_df[est_df['name'].eq('2_gen')]
est_df = est_df.pivot_table(index=run_keys, columns='dist', values='loss')

merge_df = pd.merge(
    model_df.reset_index(),
    est_df.reset_index(),
    on=run_keys,
    suffixes=('_model', '_est')
).copy()

# y(t) = train_model(t) - train_est
merge_df['y'] = merge_df['train_model'] - merge_df['train_est']

# sort within runs
df = merge_df.sort_values(run_keys + ['t']).copy()

# previous point within run
df['t_prev'] = df.groupby(run_keys)['t'].shift(1)
df['y_prev'] = df.groupby(run_keys)['y'].shift(1)

# require positive times for log interpolation
valid_time = df['t_prev'].gt(0) & df['t'].gt(0)

# downward crossing: y_prev >= 0 and y < 0
# cross = valid_time & df['y_prev'].ge(0) & df['y'].lt(-1e-4)
cross = valid_time & df['y_prev'].ge(0) & df['y'].lt(0)

# log-time coordinates
df['log_t'] = np.log(df['t'])
df['log_t_prev'] = np.log(df['t_prev'])

# interpolate linearly in log t space:
# y = y_prev + (log_t_cross - log_t_prev) * (y - y_prev)/(log_t - log_t_prev)
# solve for log_t_cross, then exponentiate
df.loc[cross, 'log_t_cross'] = (
    df.loc[cross, 'log_t_prev']
    + (0.0 - df.loc[cross, 'y_prev'])
      * (df.loc[cross, 'log_t'] - df.loc[cross, 'log_t_prev'])
      / (df.loc[cross, 'y'] - df.loc[cross, 'y_prev'])
)

df.loc[cross, 't_cross'] = np.exp(df.loc[cross, 'log_t_cross'])

t_trans = (
    df.loc[cross]
      .groupby(run_keys, as_index=False)['t_cross']
      .min()
      .rename(columns={'t_cross': 't_trans'})
)
```

```python
fig, ax = plt.subplots(figsize=(1.75, 1.5))

# subtract t_bigen from t_transience
merged = pd.merge(t_trans, t_bigen, on=run_keys)
merged['t_persistent'] = merged['t_trans'] - merged['t_bigen']

N = 8 # number of seeds per run_key
df_filtered = (
    merged
    .sort_values('seed')
    .groupby(['K', 'n'], group_keys=False)
    .head(N)
)

df_fit = (
    df_filtered
    .groupby(['K', 'n'])['t_persistent']
    .median()
    .reset_index()
)

x = df_fit['K'].values
y = df_fit['t_persistent'].values

mask = (x > 0) & (y > 0)
x, y = x[mask], y[mask]

logx = np.log(x)
logy = np.log(y)

b, loga = np.polyfit(logx, logy, 1)
a = np.exp(loga)

x_fit = np.linspace(x.min(), x.max(), 200)
y_fit = a * x_fit**b

sns.boxplot(df_filtered, x='K', y='t_persistent',
                    native_scale=True,
                    color='gray',
                    log_scale=True,
                    ax=ax)

ax.plot(x_fit, y_fit, color='black', linestyle='--')

ax.set_ylabel(r'$\tau_K$')
ax.set_xlabel(r'$K$')

ax.set_xlim(left=0)

# ax.set_ylim(bottom=10**3, top=10**6)

# plt.savefig(plot_dir / 'trans_dynamics.svg')
```

```python
def plot_loss(cur_ax, model_perf, est_perf, exp_config,
                  est_idx_list: list = list(range(4)),
                  dist: list = ['train', 'val'],
                  alpha_val: float = 0.5,
                  plot_kwargs: dict = {}):

    param_name = exp_config.param_name

    # --- estimator lines first ---
    data = est_perf
    palette = sns.color_palette()

    legend_colors = [palette[1], palette[2], palette[3], palette[0]]
    legend_names = ['1-Mem', '1-Gen', '2-Mem', '2-Gen']
    color_dict = {name: color for name, color in zip(legend_names, legend_colors)}

    # est_names = ['1_mem_last', '1_gen_last', '2_mem_last', '2_gen_last']
    est_names = ['1_mem', '1_gen', '2_mem', '2_gen']

    legend_handles = []
    for i in est_idx_list:
        for d in dist:
            est = est_names[i]
            mask = (data['name'] == est) & (data['dist'] == d)
            color = color_dict[legend_names[i]]
            cur_ax.axhline(
                y=data[mask]['loss'].unique().item(),
                label=f'{est}',
                color=color,
                alpha=alpha_val if d == 'val' else 1.0,
                **plot_kwargs
            )

        legend_handles.append(
            mlines.Line2D([], [], color=color, linestyle='-', label=f'{est}')
        )

    # --- model curves last (on top) ---
    data = model_perf
    color = 'black'

    for d in dist:
        mask = data['dist'] == d
        sns.lineplot(
            data=data[mask],
            x='t',
            y='loss',
            label=d,
            ax=cur_ax,
            color=color,
            alpha=alpha_val if d == 'val' else 1.0,
            **plot_kwargs
        )
    
    legend_handles.append(
        mlines.Line2D([], [], color=color, linestyle='-', label='Model')
    )

    cur_ax.get_legend().remove()

    cur_ax.set_title(rf'${param_name}={data[param_name].unique().item()}$')
    cur_ax.set_xlabel(r'$t$')
    cur_ax.set_ylabel('Loss')

    return legend_handles
```

```python
from base.attn_patterns import get_attn_grads
import gc

B = 2**7
repeat = 2**1

from pd.compute import naming
# from base.estimators.markov_model_ar import init_all_estimators

output = {
    'seed': [],
    'n': [],
    param_name: [],
    'pta_grad': []
}

# u.set_seed(11011)
get_task_model = u.dynamic_import(exp_config.task_model, 'get_task_model_safe')
get_model = u.dynamic_import(exp_config.model, 'get_model_safe')
for seed in exp_config.seedrange:
    for n in exp_config.nrange[:-1]:
        for param in exp_config.paramrange:
            model_dir = data_dir / naming(seed, n, param_name, param)
            state_path = model_dir / 'state'
            if state_path.exists():            
                config = u.load_config_and_task_pool(model_dir)
                task_model = get_task_model(config)

                # choose s some amount before ticl
                df = ticl.copy()
                s = df[(df['seed'] == seed) & (df['n'] == n)]['ticl'].item()
                state_idx = get_state_idx(config, state_path, s)

                state = torch.load(state_path / f'{state_idx}.pt')
                state = state['state']
                model = get_model(config)
                model.eval()
                model.load_state_dict(state)
                
                _pta_grad = []
                for _ in range(repeat):
                    x, y = task_model.get_batch(B, config.training.context_len)
                    grads = get_attn_grads(model, (x, y))
                    grads = {k: g.cpu().detach() for k, g in grads.items()}
                    pta_grad = grads[0].mean(0).diag(-1).mean()
                    _pta_grad.append(pta_grad.item())
                
                output['seed'].append(seed)
                output['n'].append(n)
                output[param_name].append(param)
                output['pta_grad'].append(sum(_pta_grad)/repeat)

output = pd.DataFrame(output)

output['pta_grad'] = output['pta_grad'].apply(lambda x: -x)
```

```python
dir = Path("/home/cg5763/data/output/proposed_marsupial")
data_dir = dir / "data"
exp_config = u.load_config(dir / 'exp_config.yaml')
print(exp_config.notes)
n_seeds = len(exp_config.seedrange)
n_n = len(exp_config.nrange)
n_params = len(exp_config.paramrange)
param_name = exp_config.param_name
model_perf = []
order_params = []
for seed in exp_config.seedrange:
    for n in exp_config.nrange:
        for param_value in exp_config.paramrange:
            try:
                _model_perf = pd.DataFrame(torch.load(data_dir / f'S{seed}_N{n}_{param_name}{param_value}' / 'log.pt'))
                _order_params = pd.read_csv(data_dir / f'S{seed}_N{n}_{param_name}{param_value}_order_params.csv')
            except FileNotFoundError:
                print(f'Missing data for seed={seed}, n={n}, {param_name}={param_value}')
            _model_perf['seed'] = seed
            _model_perf['n'] = n
            _model_perf[param_name] = param_value
            model_perf.append(_model_perf)
            order_params.append(_order_params)
model_perf = pd.concat(model_perf, ignore_index = True)
order_params = pd.concat(order_params, ignore_index = True)
```

### loss

```python
window = 20

model_perf = (
    model_perf
    .sort_values(["seed", "n", "K", "step"])  # important for rolling
)

# compute rolling average per group
model_perf["rolling"] = (
    model_perf
    .groupby(["seed", "n", "K"])["loss"]
    .transform(lambda x: x.rolling(window, min_periods=1).mean())
)
```

### t_icl

```python
ticl = (
    model_perf[model_perf["rolling"] < 2.2]
    .groupby(["seed", "n"])["step"]
    .min()
    .reset_index(name="ticl")
)
            
ticl['theory'] = ticl['n'].apply(lambda x: x/(math.log(x)))

ticl.to_csv(dir / 'ticl.csv', index=False)
```
